````python
#@title ▶️ Ejecuta esta celda SIEMPRE primero (doble clic para ver el código) { display-mode: "form" }
import os
import re
import html as html_lib
from html.parser import HTMLParser
from IPython.display import HTML, display

# 1. Montar Google Drive y ubicar la carpeta del proyecto ERP
import io
import csv
import json
import base64
import contextlib
from datetime import datetime

try:
    from google.colab import drive
    print("🔄 Conectando con Google Drive...")
    drive.mount('/content/drive')
    RUTA_DRIVE = '/content/drive/MyDrive'
except ImportError:
    RUTA_DRIVE = os.path.join(os.getcwd(), 'MyDrive')

FICHA = os.path.join(RUTA_DRIVE, 'ERP_ficha_proyecto.json')


def _ruta_del_proyecto():
    # Si ya creaste tu ficha (sesión 2), trabajamos en la carpeta ERP_<nombre de tu ERP>
    if os.path.exists(FICHA):
        with open(FICHA, encoding='utf-8') as f:
            return json.load(f)['carpeta']
    return os.path.join(RUTA_DRIVE, 'ERP_practicas')


RUTA_PROYECTO = _ruta_del_proyecto()
RUTA_SEMANA = RUTA_PROYECTO          # guardar() y leer() trabajan en la carpeta del proyecto
os.makedirs(RUTA_PROYECTO, exist_ok=True)


# 2. Guardar y leer archivos en Drive
def guardar(nombre_archivo, contenido):
    ruta = os.path.join(RUTA_SEMANA, nombre_archivo)
    with open(ruta, 'w', encoding='utf-8') as f:
        f.write(contenido)
    print(f"💾 Guardado en Drive: {ruta}")


def leer(nombre_archivo):
    ruta = os.path.join(RUTA_SEMANA, nombre_archivo)
    with open(ruta, 'r', encoding='utf-8') as f:
        return f.read()


# 3. Mostrar páginas HTML dentro de Colab (cada una en su propio marco)
def _marco(codigo_html, ancho='100%', alto=500):
    return (f'<iframe srcdoc="{html_lib.escape(codigo_html, quote=True)}" '
            f'style="width:{ancho};height:{alto}px;border:1px solid #cbd5e0;'
            f'border-radius:8px;background:#fff"></iframe>')


def mostrar(codigo_html, alto=500):
    display(HTML(_marco(codigo_html, '100%', alto)))


def lado_a_lado(html_a, html_b, titulo_a='A', titulo_b='B', alto=320):
    columnas = ''
    for titulo, pagina in ((titulo_a, html_a), (titulo_b, html_b)):
        columnas += (f'<div style="flex:1;min-width:280px">'
                     f'<div style="font:bold 14px sans-serif;margin-bottom:4px">{titulo}</div>'
                     f'{_marco(pagina, "100%", alto)}</div>')
    display(HTML(f'<div style="display:flex;gap:16px;flex-wrap:wrap">{columnas}</div>'))


def con_css(codigo_html, css):
    # Inserta una hoja de estilos <style> justo antes de </head>
    return codigo_html.replace('</head>', f'<style>\n{css}\n</style>\n</head>', 1)


def comparar_pantallas(codigo_html, anchos=(360, 768, 1280), alto=560):
    marcos = ''
    for ancho in anchos:
        escala = min(1, 380 / ancho)
        icono = '📱' if ancho < 600 else ('📟' if ancho < 1024 else '🖥️')
        marcos += (f'<div style="flex:none">'
                   f'<div style="font:bold 13px sans-serif;margin-bottom:4px">{icono} {ancho}px</div>'
                   f'<div style="width:{ancho * escala:.0f}px;height:{alto}px;overflow:hidden">'
                   f'<div style="width:{ancho}px;transform:scale({escala});transform-origin:0 0">'
                   f'{_marco(codigo_html, f"{ancho}px", round(alto / escala))}</div></div></div>')
    display(HTML(f'<div style="display:flex;gap:16px;overflow-x:auto;padding:8px 0">{marcos}</div>'))


# 4. Analizador de HTML: radiografía semántica y revisor automático
REGIONES = {
    'header': 'banner (encabezado)',
    'nav': 'navigation (navegación)',
    'main': 'main (contenido principal)',
    'section': 'region (sección temática)',
    'article': 'article (contenido independiente)',
    'aside': 'complementary (complementario)',
    'footer': 'contentinfo (pie de página)',
    'form': 'form (formulario)',
}


class _Analizador(HTMLParser):
    def __init__(self):
        super().__init__()
        self.etiquetas = []
        self.doctype = ''
        self.titulo = ''
        self.niveles = []
        self.labels_for = set()
        self.controles = []
        self.botones_vacios = 0
        self.mapa = []
        self._prof = 0
        self._en_title = False
        self._labels_abiertos = 0
        self._boton = None
        self._encabezado = None

    def handle_decl(self, decl):
        self.doctype = decl

    def handle_starttag(self, tag, attrs):
        a = {k: (v or '') for k, v in attrs}
        self.etiquetas.append((tag, a))
        if tag == 'title':
            self._en_title = True
        if tag == 'label':
            self._labels_abiertos += 1
            if a.get('for'):
                self.labels_for.add(a['for'])
        if tag in ('input', 'select', 'textarea'):
            self.controles.append((tag, a, self._labels_abiertos > 0))
        if tag == 'button':
            self._boton = 'x' if a.get('aria-label') else ''
        if tag in REGIONES:
            self.mapa.append((self._prof, f"🏷️  <{tag}>  →  {REGIONES[tag]}"))
            self._prof += 1
        if re.fullmatch(r'h[1-6]', tag):
            self._encabezado = [int(tag[1]), '']
            self.niveles.append(int(tag[1]))

    def handle_endtag(self, tag):
        if tag == 'title':
            self._en_title = False
        if tag == 'label':
            self._labels_abiertos = max(0, self._labels_abiertos - 1)
        if tag == 'button' and self._boton is not None:
            if not self._boton.strip():
                self.botones_vacios += 1
            self._boton = None
        if tag in REGIONES:
            self._prof = max(0, self._prof - 1)
        if self._encabezado and tag == f'h{self._encabezado[0]}':
            nivel, texto = self._encabezado
            self.mapa.append((self._prof, f"📑 h{nivel}: {' '.join(texto.split())}"))
            self._encabezado = None

    def handle_data(self, data):
        if self._en_title:
            self.titulo += data
        if self._boton is not None:
            self._boton += data
        if self._encabezado:
            self._encabezado[1] += data


def _analizar(codigo_html):
    analizador = _Analizador()
    analizador.feed(codigo_html)
    return analizador


def radiografia(codigo_html, nombre='la página'):
    a = _analizar(codigo_html)
    print(f"🔎 Radiografía de {nombre}: lo que entiende un navegador, Google o un lector de pantalla")
    print("-" * 75)
    if not a.mapa:
        print("⚠️  No encontré regiones ni encabezados.")
        print("    Para un lector de pantalla, esta página es un solo bloque de texto sin estructura.")
    for profundidad, linea in a.mapa:
        print("    " * profundidad + linea)
    etiquetas = [t for t, _ in a.etiquetas]
    semanticas = sum(etiquetas.count(t) for t in REGIONES)
    print("-" * 75)
    print(f"📊 Etiquetas semánticas: {semanticas}   |   <div>: {etiquetas.count('div')}")


def revisar_html(codigo_html, nombre='la maqueta'):
    a = _analizar(codigo_html)
    etiquetas = [t for t, _ in a.etiquetas]
    de_tipo = lambda tag: [attrs for t, attrs in a.etiquetas if t == tag]
    metas = de_tipo('meta')
    raiz = de_tipo('html')
    semanticas = sum(etiquetas.count(t) for t in REGIONES)

    sin_label = []
    for tag, attrs, dentro_de_label in a.controles:
        if attrs.get('type') in ('hidden', 'submit', 'button', 'reset', 'image'):
            continue
        tiene = (dentro_de_label or attrs.get('id') in a.labels_for
                 or attrs.get('aria-label') or attrs.get('aria-labelledby'))
        if not tiene:
            sin_label.append(attrs.get('id') or attrs.get('name') or attrs.get('placeholder') or tag)

    botones_falsos = [x.get('class') for t, x in a.etiquetas
                      if t in ('div', 'span') and re.search(r'bot[oó]n|btn', x.get('class', ''))]
    saltos = [f"h{p}→h{n}" for p, n in zip(a.niveles, a.niveles[1:]) if n > p + 1]
    imgs_sin_alt = [x.get('src', '?') for x in de_tipo('img') if 'alt' not in x]

    pruebas = [
        (a.doctype.lower().startswith('doctype html'),
         'Declara <!DOCTYPE html>', 'Escribe <!DOCTYPE html> en la primera línea.'),
        (bool(raiz) and bool(raiz[0].get('lang')),
         'La etiqueta <html> indica el idioma', 'Usa <html lang="es"> para que el lector de pantalla pronuncie en español.'),
        (any('charset' in m for m in metas),
         'Declara la codificación (meta charset)', 'Agrega <meta charset="UTF-8"> para que se vean bien los acentos y la ñ.'),
        (any(m.get('name') == 'viewport' for m in metas),
         'Tiene meta viewport (lista para celular)', 'Agrega <meta name="viewport" content="width=device-width, initial-scale=1.0">.'),
        (bool(a.titulo.strip()),
         'Tiene un <title> descriptivo', 'Escribe un <title> con el nombre de tu app.'),
        (etiquetas.count('h1') == 1,
         'Usa exactamente un <h1>', f'Hay {etiquetas.count("h1")} <h1>; deja solo uno, con el nombre de la app o de la pantalla.'),
        (bool(a.niveles) and not saltos,
         'Los encabezados siguen un orden (sin saltos)', 'Ordena los títulos h1 → h2 → h3. ' + (f'Saltos: {", ".join(saltos)}' if saltos else 'No hay encabezados.')),
        ('header' in etiquetas,
         'Tiene <header>', 'Envuelve el logo y el título de tu app en <header>.'),
        ('nav' in etiquetas,
         'Tiene <nav>', 'Pon el menú principal dentro de <nav>.'),
        (etiquetas.count('main') == 1,
         'Tiene un solo <main>', f'Hay {etiquetas.count("main")} <main>; el contenido principal va en un único <main>.'),
        ('footer' in etiquetas,
         'Tiene <footer>', 'Agrega un <footer> con los créditos de tu proyecto.'),
        (not imgs_sin_alt,
         'Todas las imágenes tienen texto alternativo (alt)', f'Imágenes sin alt: {", ".join(imgs_sin_alt)}'),
        (not sin_label,
         'Cada campo del formulario tiene <label>', f'Campos sin etiqueta: {", ".join(sin_label)}. El placeholder NO sustituye a <label>.'),
        (a.botones_vacios == 0,
         'Todos los botones tienen texto', 'Escribe un texto dentro de cada <button>, o agrega aria-label.'),
        (not botones_falsos,
         'Los botones son <button> (no <div> disfrazados)', 'Un <div class="boton"> no se puede activar con teclado ni lo anuncia el lector de pantalla. Usa <button>.'),
        ('TODO' not in codigo_html,
         'No quedan TODO pendientes', f'Quedan {codigo_html.count("TODO")} TODO por completar con los datos de tu proyecto.'),
        ('table' not in etiquetas or 'th' in etiquetas,
         'Las tablas tienen encabezados <th>', 'Usa <th scope="col"> en la primera fila de tus tablas.'),
        (etiquetas.count('div') <= semanticas,
         'Sin "divitis" (más etiquetas semánticas que <div>)', f'Hay {etiquetas.count("div")} <div> y {semanticas} etiquetas semánticas. ¿Cuáles <div> podrían ser header, nav, section…?'),
    ]

    aprobadas = sum(ok for ok, _, _ in pruebas)
    print(f"🧪 Revisión automática de {nombre}")
    print("=" * 75)
    for ok, criterio, consejo in pruebas:
        print(f"✅ {criterio}" if ok else f"❌ {criterio}\n      💡 {consejo}")
    print("=" * 75)
    porcentaje = round(100 * aprobadas / len(pruebas))
    medalla = '🏆' if porcentaje == 100 else ('👍' if porcentaje >= 80 else '🔧')
    print(f"{medalla} Resultado: {aprobadas}/{len(pruebas)} criterios ({porcentaje} %)")
    return porcentaje


# 5. Predicciones: ¿qué imprimirá este código?
def predecir(codigo, mi_prediccion):
    if not str(mi_prediccion).strip():
        print("✏️ Primero escribe tu predicción en MI_PREDICCION y vuelve a ejecutar.")
        return
    salida = io.StringIO()
    with contextlib.redirect_stdout(salida):
        exec(codigo, {})
    real = salida.getvalue().strip()
    if ' '.join(str(mi_prediccion).split()) == ' '.join(real.split()):
        print(f"✅ ¡Acertaste! El código imprime: {real}")
    else:
        print(f"🤔 Tu predicción:  {mi_prediccion}")
        print(f"   Resultado real: {real}")
        print("   💬 Coméntalo con tu pareja: ¿en qué línea cambió lo que esperabas?")


# 6. Autoverificador: cada prueba es (descripción, función que devuelve True/False, pista)
def autoverificar(titulo, pruebas, meta=None):
    aprobadas = 0
    print(f"✅ Autoverificador · {titulo}")
    print("=" * 75)
    for descripcion, prueba, pista in pruebas:
        try:
            ok = bool(prueba())
            error = ''
        except Exception as e:
            ok = False
            error = f" ({type(e).__name__}: {e})"
        aprobadas += ok
        print(f"✅ {descripcion}" if ok else f"❌ {descripcion}{error}\n      💡 {pista}")
    print("=" * 75)
    porcentaje = round(100 * aprobadas / len(pruebas))
    meta_txt = f" · meta: {meta}" if meta else ''
    medalla = '🏆' if porcentaje == 100 else ('👍' if porcentaje >= 75 else '🔧')
    print(f"{medalla} Resultado: {aprobadas}/{len(pruebas)} ({porcentaje} %){meta_txt}")
    return porcentaje


# 7. Guardar y leer el trabajo del proyecto (JSON en la carpeta docs/)
def guardar_json(nombre, datos):
    ruta = os.path.join(RUTA_PROYECTO, 'docs', nombre)
    os.makedirs(os.path.dirname(ruta), exist_ok=True)
    with open(ruta, 'w', encoding='utf-8') as f:
        json.dump(datos, f, ensure_ascii=False, indent=2)
    print(f"💾 Guardado en tu proyecto: docs/{nombre}")


def leer_json(nombre, por_defecto=None):
    ruta = os.path.join(RUTA_PROYECTO, 'docs', nombre)
    if not os.path.exists(ruta):
        return por_defecto
    with open(ruta, encoding='utf-8') as f:
        return json.load(f)


# 8. Bitácoras: tickets de salida y uso de IA
def _anotar(archivo, encabezado, fila):
    ruta = os.path.join(RUTA_PROYECTO, 'bitacora', archivo)
    os.makedirs(os.path.dirname(ruta), exist_ok=True)
    nuevo = not os.path.exists(ruta)
    with open(ruta, 'a', encoding='utf-8', newline='') as f:
        escritor = csv.writer(f)
        if nuevo:
            escritor.writerow(encabezado)
        escritor.writerow(fila)


def ticket_salida(sesion, aprendi, me_bloqueo):
    if not aprendi.strip():
        print("✏️ Escribe qué aprendiste hoy antes de registrar tu ticket.")
        return
    _anotar('tickets_salida.csv', ['fecha', 'sesion', 'aprendi', 'me_bloqueo'],
            [f"{datetime.now():%Y-%m-%d %H:%M}", sesion, aprendi.strip(), me_bloqueo.strip()])
    print(f"🎟️ Ticket de la sesión {sesion} registrado en bitacora/tickets_salida.csv")


def registrar_ia(pregunta, respuesta, que_acepte, como_verifique):
    _anotar('bitacora_ia.csv', ['fecha', 'pregunta', 'respuesta', 'que_acepte', 'como_verifique'],
            [f"{datetime.now():%Y-%m-%d %H:%M}", pregunta, respuesta, que_acepte, como_verifique])
    print("🤖 Uso de IA registrado en bitacora/bitacora_ia.csv")


# 9. Ver el árbol de carpetas del proyecto
def arbol(ruta=None, prefijo=''):
    ruta = ruta or RUTA_PROYECTO
    if not prefijo:
        print(f"📁 {os.path.basename(ruta)}/")
    elementos = sorted(os.listdir(ruta))
    for i, nombre in enumerate(elementos):
        ultimo = i == len(elementos) - 1
        completo = os.path.join(ruta, nombre)
        print(f"{prefijo}{'└── ' if ultimo else '├── '}{'📁 ' if os.path.isdir(completo) else '📄 '}{nombre}")
        if os.path.isdir(completo):
            arbol(completo, prefijo + ('    ' if ultimo else '│   '))


# 10. Herramientas de diseño (semana 2)
HEURISTICOS = {
    1: "Visibilidad del estado del sistema",
    2: "Coincidencia entre el sistema y el mundo real",
    3: "Control y libertad del usuario",
    4: "Consistencia y estándares",
    5: "Prevención de errores",
    6: "Reconocer en lugar de recordar",
    7: "Flexibilidad y eficiencia de uso",
    8: "Diseño estético y minimalista",
    9: "Ayudar a reconocer, diagnosticar y recuperarse de errores",
    10: "Ayuda y documentación",
}


def _es_color(color):
    return isinstance(color, str) and re.fullmatch(r'#[0-9a-fA-F]{6}', color.strip()) is not None


def _contraste_ref(color_1, color_2):
    # Cálculo de referencia (lo usan los ejemplos y los autoverificadores)
    def luz(valor):
        c = valor / 255
        return c / 12.92 if c <= 0.03928 else ((c + 0.055) / 1.055) ** 2.4

    def luminancia_ref(h):
        h = h.strip().lstrip('#')
        r, g, b = (int(h[i:i + 2], 16) for i in (0, 2, 4))
        return 0.2126 * luz(r) + 0.7152 * luz(g) + 0.0722 * luz(b)

    claro, oscuro = sorted((luminancia_ref(color_1), luminancia_ref(color_2)), reverse=True)
    return (claro + 0.05) / (oscuro + 0.05)


def _tarjetas_color(items):
    # items: (color_texto, color_fondo, etiqueta, razon, minimo)
    tarjetas = ''
    for fg, bg, etiqueta, razon, minimo in items:
        ok = razon >= minimo
        tarjetas += (
            f'<div style="flex:1;min-width:200px;max-width:260px;border:1px solid #cbd5e0;border-radius:10px;'
            f'overflow:hidden;font-family:system-ui,sans-serif;background:#fff">'
            f'<div style="background:{bg};color:{fg};padding:14px;font-size:18px;font-weight:bold">Aa · Venta guardada</div>'
            f'<div style="padding:8px 12px;font-size:13px;color:#2d3748">'
            f'<b>{html_lib.escape(etiqueta)}</b><br>{fg} sobre {bg}<br>'
            f'Razón: <b>{razon:.2f} : 1</b><br>'
            f'{"✅ Cumple" if ok else "❌ No cumple"} (mínimo {minimo:g} : 1)</div></div>')
    display(HTML(f'<div style="display:flex;gap:12px;flex-wrap:wrap;padding:6px 0">{tarjetas}</div>'))


def mostrar_pares(pares, texto_grande=False):
    # pares: lista de (color_del_texto, color_de_fondo, etiqueta)
    minimo = 3 if texto_grande else 4.5
    _tarjetas_color([(fg, bg, etiqueta, _contraste_ref(fg, bg), minimo) for fg, bg, etiqueta in pares])


def mostrar_paleta(paleta, razon=None):
    # Muestra tu paleta y calcula la razón de contraste con TU función razon_contraste
    # (si no le pasas ninguna, usa la función de referencia del curso).
    razon = razon or _contraste_ref
    combinaciones = [
        ('texto', 'fondo', 'Texto sobre el fondo', 4.5),
        ('texto_sobre_primario', 'primario', 'Texto dentro de los botones', 4.5),
        ('primario', 'fondo', 'Botón contra el fondo (componente)', 3),
    ]
    if 'alerta' in paleta:
        combinaciones.append(('alerta', 'fondo', 'Color de alerta sobre el fondo', 4.5))
    faltan = [c for par in combinaciones for c in par[:2] if not _es_color(paleta.get(c))]
    if faltan:
        print(f"✏️ Revisa estos colores de tu paleta (formato #rrggbb): {', '.join(sorted(set(faltan)))}")
        return
    try:
        items = [(paleta[a], paleta[b], etiqueta, float(razon(paleta[a], paleta[b])), minimo)
                 for a, b, etiqueta, minimo in combinaciones]
    except Exception as error:
        print(f"⚠️ Tu función razon_contraste todavía no funciona ({type(error).__name__}). Completa el reto Base primero.")
        return
    _tarjetas_color(items)


def diagrama_flujo(titulo, flujo):
    # flujo = {"entrada": [...], "proceso": [...], "salida": [...]}
    columnas = (('entrada', '📥 Entradas', '#ebf8ff', '#2b6cb0'),
                ('proceso', '⚙️ Proceso', '#faf5ff', '#6b46c1'),
                ('salida', '📤 Salidas', '#f0fff4', '#276749'))
    cajas = []
    for clave, rotulo, fondo, borde in columnas:
        elementos = ''.join(f'<li>{html_lib.escape(str(x))}</li>' for x in flujo.get(clave, [])) or '<li><i>(vacío)</i></li>'
        cajas.append(f'<div style="flex:1;min-width:190px;background:{fondo};border:2px solid {borde};border-radius:10px;padding:10px 14px">'
                     f'<div style="font-weight:bold;color:{borde};margin-bottom:6px">{rotulo}</div>'
                     f'<ul style="margin:0;padding-left:1.1rem;font-size:14px;color:#1a202c">{elementos}</ul></div>')
    flecha = '<div style="align-self:center;font-size:26px;color:#4a5568">➜</div>'
    display(HTML(f'<div style="font-family:system-ui,sans-serif"><div style="font-weight:bold;font-size:16px;margin-bottom:8px">'
                 f'🔄 {html_lib.escape(titulo)}</div><div style="display:flex;gap:8px;flex-wrap:wrap">'
                 f'{flecha.join(cajas)}</div></div>'))


def clics_desde(enlaces, inicio="Inicio"):
    # Recorrido por niveles: ¿cuántos clics hacen falta para llegar a cada pantalla?
    clics = {}
    if inicio not in enlaces:
        return clics
    clics[inicio] = 0
    cola = [inicio]
    while cola:
        actual = cola.pop(0)
        for destino in enlaces.get(actual, []):
            if destino in enlaces and destino not in clics:
                clics[destino] = clics[actual] + 1
                cola.append(destino)
    return clics


def _problemas_navegacion(enlaces, inicio="Inicio", max_clics=3, minimo=5):
    clics = clics_desde(enlaces, inicio)
    pantallas = list(enlaces)
    return {
        'sin_inicio': inicio not in enlaces,
        'pocas': len(pantallas) < minimo,
        'rotos': [f"{o} → {d}" for o, ds in enlaces.items() for d in ds if d not in enlaces],
        'aisladas': [p for p in pantallas if p not in clics],
        'lejanas': [f"{p} ({n} clics)" for p, n in clics.items() if n > max_clics],
        'callejones': [p for p in pantallas if not enlaces[p]],
        'clics': clics,
    }


def arbol_navegacion(enlaces, inicio="Inicio"):
    clics = clics_desde(enlaces, inicio)
    hijos = {p: [] for p in enlaces}
    for pantalla in clics:                      # cada pantalla cuelga de la primera que la descubrió
        for destino in enlaces.get(pantalla, []):
            if destino in clics and clics[destino] == clics[pantalla] + 1 and not any(destino in h for h in hijos.values()):
                hijos[pantalla].append(destino)
    print("🌳 Mapa de navegación (cada nivel = 1 clic desde Inicio)")
    print("-" * 75)
    if inicio not in enlaces:
        print(f"⚠️ No existe la pantalla «{inicio}». Todo mapa empieza en Inicio.")
        return

    def imprimir(pantalla, prefijo='', ultimo=True, raiz=False):
        marca = '🏠 ' if raiz else ''
        etiqueta = '' if raiz else f"  ({clics[pantalla]} clic{'s' if clics[pantalla] != 1 else ''})"
        print(f"{prefijo}{'' if raiz else ('└── ' if ultimo else '├── ')}{marca}{pantalla}{etiqueta}")
        siguientes = hijos[pantalla]
        for i, hijo in enumerate(siguientes):
            extra = '' if raiz else ('    ' if ultimo else '│   ')
            imprimir(hijo, prefijo + extra, i == len(siguientes) - 1)

    imprimir(inicio, raiz=True)
    problemas = _problemas_navegacion(enlaces, inicio)
    if problemas['aisladas']:
        print(f"🚫 Sin camino desde {inicio}: {', '.join(problemas['aisladas'])}")
    if problemas['rotos']:
        print(f"⚠️ Enlaces a pantallas que no existen: {', '.join(problemas['rotos'])}")


def verificar_navegacion(enlaces, inicio="Inicio", max_clics=3, minimo=5, nombre="mapa de navegación"):
    p = _problemas_navegacion(enlaces, inicio, max_clics, minimo)
    pruebas = [
        (not p['sin_inicio'], f'Existe la pantalla «{inicio}»', f'Agrega la pantalla «{inicio}» al diccionario.'),
        (not p['pocas'], f'Tiene {minimo} o más pantallas', f'Solo hay {len(enlaces)} pantallas.'),
        (not p['rotos'], 'Todos los enlaces llevan a una pantalla que existe', f"Enlaces rotos: {', '.join(p['rotos'])}. Crea la pantalla o corrige el nombre."),
        (not p['aisladas'], 'Ninguna pantalla está aislada (todas se alcanzan desde Inicio)', f"Nadie lleva a: {', '.join(p['aisladas'])}. Agrega un botón que lleve hasta ellas."),
        (not p['lejanas'], f'Toda pantalla está a {max_clics} clics o menos', f"Muy lejos: {', '.join(p['lejanas'])}. Acércala con un acceso directo desde Inicio."),
        (not p['callejones'], 'Ninguna pantalla es un callejón sin salida', f"Sin salida: {', '.join(p['callejones'])}. Toda pantalla necesita un botón para regresar o seguir (Nielsen #3)."),
    ]
    aprobadas = sum(ok for ok, _, _ in pruebas)
    print(f"🧭 Verificación de navegación · {nombre}")
    print("=" * 75)
    for ok, criterio, consejo in pruebas:
        print(f"✅ {criterio}" if ok else f"❌ {criterio}\n      💡 {consejo}")
    print("=" * 75)
    porcentaje = round(100 * aprobadas / len(pruebas))
    medalla = '🏆' if porcentaje == 100 else ('👍' if porcentaje >= 80 else '🔧')
    print(f"{medalla} Resultado: {aprobadas}/{len(pruebas)} criterios ({porcentaje} %)")
    return porcentaje


def guion_prototipo(enlaces):
    # Convierte tu mapa de navegación en la lista de diapositivas y botones que debes crear en Slides
    numeros = {pantalla: i for i, pantalla in enumerate(enlaces, 1)}
    print("🖼️ Guion de tu prototipo en Google Slides (una diapositiva por pantalla)")
    print("=" * 75)
    for pantalla, destinos in enlaces.items():
        print(f"📄 Diapositiva {numeros[pantalla]}: {pantalla}")
        for destino in destinos:
            if destino in numeros:
                print(f"     🔘 Botón «{destino}»  →  diapositiva {numeros[destino]}")
            else:
                print(f"     ⚠️ Botón «{destino}»  →  esa pantalla no existe en tu mapa")
        if not destinos:
            print("     ⚠️ Esta pantalla no tiene botones: ¿cómo sale la persona de aquí?")
    print("=" * 75)
    print(f"Total: {len(numeros)} diapositivas. Ordénalas tal como salen aquí para que los números coincidan.")


print("=" * 75)
print(f"📁 Carpeta del proyecto:\n{RUTA_PROYECTO}")
print("🧰 Herramientas cargadas: predecir, autoverificar, guardar_json, leer_json, ticket_salida,")
print("   registrar_ia, arbol, mostrar, lado_a_lado, comparar_pantallas, radiografia, revisar_html")
print("   NUEVAS (semana 2): diagrama_flujo, arbol_navegacion, verificar_navegacion, guion_prototipo,")
print("   clics_desde, mostrar_pares, mostrar_paleta, HEURISTICOS")
print("=" * 75)
````

## 🔄 A · Flujo de datos: qué entra, qué pasa y qué sale
Antes de dibujar una pantalla hay que saber **qué datos entran, qué hace el sistema con ellos y qué debe salir**. Es el modelo **Entrada → Proceso → Salida**. Su versión profesional es el *diagrama de flujo de datos* (DFD), que la norma **ISO/IEC/IEEE 29148** menciona como una forma de mostrar cómo se relacionan las funciones y los datos *desde el punto de vista del usuario*.

| Parte | Pregunta | Ejemplo: “registrar una venta” |
|---|---|---|
| 📥 **Entrada** | ¿Qué escribe, elige o escanea la persona? | Producto, cantidad, cliente (si es fiado), forma de pago |
| ⚙️ **Proceso** | ¿Qué calcula, valida o guarda el sistema? | Busca el precio, calcula el total, descuenta el inventario, guarda la venta |
| 📤 **Salida** | ¿Qué ve o recibe la persona? | Ticket con el total, mensaje “Venta guardada”, alerta de stock bajo |

> 💡 **Cada pantalla de tu ERP existe para que algo entre, se procese o salga.** Si una pantalla no hace ninguna de las tres cosas, sobra.

## 🗺️ B · Mapa de navegación: cómo se llega de una pantalla a otra
Un **mapa de navegación** (en inglés *screen flow*) es un dibujo con una caja por pantalla y una flecha por cada botón o enlace. Alistair Cockburn (*Agile Software Development*) describe tres niveles de precisión para describir una interfaz:

| Nivel | Qué muestra | ¿Cuándo lo hacemos? |
|---|---|---|
| 1 · Baja precisión | Diagrama de flujo de pantallas: para qué sirve cada una y cómo se enlazan | **Sesión 2** (tu mapa) |
| 2 · Precisión media | Campos, reglas y botones de cada pantalla | **Sesión 3** (tu prototipo) |
| 3 · Alta precisión | El código final de la interfaz | Semanas 3 y 4 (HTML y CSS) |

**Tres reglas para un buen mapa**
1. Desde **Inicio** se llega a cualquier pantalla en **3 clics o menos**.
2. **Ninguna pantalla es un callejón sin salida:** siempre hay un botón para regresar o seguir *(Nielsen #3: control y libertad)*.
3. El menú vive en el mismo lugar en todas las pantallas *(Nielsen #4: consistencia)*.

En Python, el mapa es un **diccionario**: cada llave es una pantalla y su valor es la lista de pantallas a las que se puede ir desde ella.

## 🎨 C · Teoría de diseño de pantallas
Una pantalla bien diseñada se entiende **en 3 segundos**. Estos cinco principios te sirven para cualquier ERP:

| Principio | Qué significa | Regla práctica |
|---|---|---|
| 🥇 **Jerarquía visual** | Lo importante se ve primero | Un solo título por pantalla (`h1`), una acción principal destacada (botón grande y de color) y lo secundario más discreto. Los títulos siguen el orden `h1 → h2 → h3` |
| 🎨 **Color** | Pocos colores, cada uno con una función | Fondo claro, texto oscuro, **1 color primario** para acciones y **1 color de alerta**. Guíate por la proporción 60-30-10 (60 % neutro, 30 % secundario, 10 % acento). Define los colores como variables CSS en `:root` |
| 🔤 **Tipografía** | Que se lea a la primera | Una sola familia, cuerpo de **16 px o más** y medidas relativas (`rem`, `em`) para que el zoom del usuario funcione *(MDN, diseño receptivo)* |
| 📐 **Espaciado** | El aire agrupa | Lo relacionado va junto; los grupos, separados. Usa `box-sizing: border-box` para que el relleno no desborde |
| 🔁 **Consistencia** | Lo mismo se ve y se comporta igual | Mismo menú, mismos colores y mismos botones en todas las pantallas *(Nielsen #4)* |

## ♿ D · Accesibilidad con WCAG: diseñar para todas las personas
La **accesibilidad web** significa que personas con discapacidad visual, auditiva, motriz o cognitiva —y también quien usa el celular bajo el sol, con un dedo lastimado o con internet lento— pueden **percibir, entender, navegar e interactuar** con tu sistema *(W3C WAI, Introducción a la accesibilidad web)*. Las pautas **WCAG** se organizan en cuatro principios:

| Principio | Pregunta | Ejemplo en tu ERP |
|---|---|---|
| 👁️ **Perceptible** | ¿Se puede ver u oír? | Texto con buen contraste; imágenes con `alt` |
| 🖐️ **Operable** | ¿Se puede usar? | Todo funciona con teclado; botones grandes |
| 🧠 **Comprensible** | ¿Se entiende? | Etiquetas claras; mensajes de error que explican |
| 🔧 **Robusto** | ¿Funciona con distintas herramientas? | HTML semántico que lee un lector de pantalla |

**Cinco comprobaciones que haremos en el prototipo**

| Comprobación | Criterio WCAG | Regla |
|---|---|---|
| 🌗 **Contraste** | 1.4.3 y 1.4.11 | Texto normal: **4.5 : 1** o más. Texto grande (24 px o más, o 18.66 px en negrita): **3 : 1**. Bordes de botones e íconos: **3 : 1** |
| 🖼️ **Texto alternativo** | 1.1.1 | Toda imagen que informa lleva `alt` |
| ⌨️ **Teclado** | 2.1.1 | Todo se puede hacer sin mouse: usa `<button>` y `<a>`, no `<div>` disfrazados |

| 🚦 **No solo color** | 1.4.1 | Si algo está mal, no lo indiques solo con rojo: agrega texto o ícono |

### 🌗 El contraste se puede calcular
La **razón de contraste** va de 1 : 1 (idéntico) a 21 : 1 (negro sobre blanco):

```
razón = (L1 + 0.05) / (L2 + 0.05)         L1 = luminancia del color más claro, L2 = la del más oscuro
L = 0.2126·R + 0.7152·G + 0.0722·B        (R, G y B ya convertidos a escala lineal)
canal lineal: c = valor / 255
              si c ≤ 0.03928 → c / 12.92
              si no          → ((c + 0.055) / 1.055) ** 2.4
```
El verde pesa más (0.7152) porque el ojo humano es más sensible a él. **En la sesión 2 usarás esta calculadora (ya viene hecha en `mostrar_paleta`) para elegir los colores de tu ERP;** quien termine antes puede programarla como ⭐ Bonus. Por ahora, observa el resultado en estos ejemplos:

## 📱 E · Mobile-first: se diseña primero para el celular
La mayoría de tus usuarios abrirá el ERP en un celular. **Mobile-first** (idea de Luke Wroblewski, que se une al *diseño web adaptable* o *responsive* de Ethan Marcotte) es escribir primero los estilos base para pantallas **estrechas** y después **agregar** estilos para pantallas más grandes con `@media (min-width: …)` *(MDN, diseño receptivo)*.

| Pieza | Para qué sirve |
|---|---|
| `<meta name="viewport" content="width=device-width, initial-scale=1.0">` | Hace que el celular use su ancho real y no finja ser una pantalla de 960 px |
| Estilos base sin `@media` | Una columna, botones grandes, texto de 16 px o más |
| `@media (min-width: 768px) { … }` | Cuando hay espacio, agrega columnas |
| Flexbox y Grid con `fr` | Reparten el espacio sin medidas fijas |
| `img { max-width: 100%; }` | Las imágenes nunca se salen de la pantalla |

### 👩‍🏫 Mini demo: el contraste paso a paso
Calculamos con código el contraste del texto **azul marino `#1a365d`** sobre un **fondo blanco `#ffffff`**. Ejecuta la celda y sigue cada paso (en el reto no tendrás que programarlo: `mostrar_paleta()` lo hace por ti):

````python
texto, fondo = "#1a365d", "#ffffff"

# 1) De "#1a365d" a tres números entre 0 y 255
hexa = texto.lstrip("#")
r, g, b = int(hexa[0:2], 16), int(hexa[2:4], 16), int(hexa[4:6], 16)
print("1) RGB del texto:", (r, g, b))

# 2) Cada canal pasa a escala lineal (dos tramos)
canales = []
for valor in (r, g, b):
    c = valor / 255
    if c <= 0.03928:
        canales.append(c / 12.92)
    else:
        canales.append(((c + 0.055) / 1.055) ** 2.4)
print("2) Canales en escala lineal:", [round(x, 4) for x in canales])

# 3) Luminancia: el verde pesa más que el rojo y que el azul
L_texto = 0.2126 * canales[0] + 0.7152 * canales[1] + 0.0722 * canales[2]
print("3) Luminancia del texto:", round(L_texto, 4))

# 4) El blanco tiene luminancia 1.0. La razón usa la mayor arriba y la menor abajo
L_fondo = 1.0
razon = (L_fondo + 0.05) / (L_texto + 0.05)
print(f"4) Razón de contraste: {razon:.2f} : 1")
print("5) ¿Cumple 4.5 : 1?", "✅ Sí" if razon >= 4.5 else "❌ No")

mostrar_pares([(texto, fondo, "Mi ejemplo")])
````
