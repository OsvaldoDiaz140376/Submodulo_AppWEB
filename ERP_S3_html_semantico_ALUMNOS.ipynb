````python
#@title ▶️ Ejecuta esta celda SIEMPRE primero (doble clic para ver el código) { display-mode: "form" }
import os
import re
import html as html_lib
from html.parser import HTMLParser
from IPython.display import HTML, display

# 1. Montar Google Drive y ubicar la carpeta del proyecto ERP
import io
import csv
import json
import base64
import contextlib
from datetime import datetime

try:
    from google.colab import drive
    print("🔄 Conectando con Google Drive...")
    drive.mount('/content/drive')
    RUTA_DRIVE = '/content/drive/MyDrive'
except ImportError:
    RUTA_DRIVE = os.path.join(os.getcwd(), 'MyDrive')

FICHA = os.path.join(RUTA_DRIVE, 'ERP_ficha_proyecto.json')


def _ruta_del_proyecto():
    # Si ya creaste tu ficha (sesión 2), trabajamos en la carpeta ERP_<nombre de tu ERP>
    if os.path.exists(FICHA):
        with open(FICHA, encoding='utf-8') as f:
            return json.load(f)['carpeta']
    return os.path.join(RUTA_DRIVE, 'ERP_practicas')


RUTA_PROYECTO = _ruta_del_proyecto()
RUTA_SEMANA = RUTA_PROYECTO          # guardar() y leer() trabajan en la carpeta del proyecto
os.makedirs(RUTA_PROYECTO, exist_ok=True)


# 2. Guardar y leer archivos en Drive
def guardar(nombre_archivo, contenido):
    ruta = os.path.join(RUTA_SEMANA, nombre_archivo)
    with open(ruta, 'w', encoding='utf-8') as f:
        f.write(contenido)
    print(f"💾 Guardado en Drive: {ruta}")


def leer(nombre_archivo):
    ruta = os.path.join(RUTA_SEMANA, nombre_archivo)
    with open(ruta, 'r', encoding='utf-8') as f:
        return f.read()


# 3. Mostrar páginas HTML dentro de Colab (cada una en su propio marco)
def _marco(codigo_html, ancho='100%', alto=500):
    return (f'<iframe srcdoc="{html_lib.escape(codigo_html, quote=True)}" '
            f'style="width:{ancho};height:{alto}px;border:1px solid #cbd5e0;'
            f'border-radius:8px;background:#fff"></iframe>')


def mostrar(codigo_html, alto=500):
    display(HTML(_marco(codigo_html, '100%', alto)))


def lado_a_lado(html_a, html_b, titulo_a='A', titulo_b='B', alto=320):
    columnas = ''
    for titulo, pagina in ((titulo_a, html_a), (titulo_b, html_b)):
        columnas += (f'<div style="flex:1;min-width:280px">'
                     f'<div style="font:bold 14px sans-serif;margin-bottom:4px">{titulo}</div>'
                     f'{_marco(pagina, "100%", alto)}</div>')
    display(HTML(f'<div style="display:flex;gap:16px;flex-wrap:wrap">{columnas}</div>'))


def con_css(codigo_html, css):
    # Inserta una hoja de estilos <style> justo antes de </head>
    return codigo_html.replace('</head>', f'<style>\n{css}\n</style>\n</head>', 1)


def comparar_pantallas(codigo_html, anchos=(360, 768, 1280), alto=560):
    marcos = ''
    for ancho in anchos:
        escala = min(1, 380 / ancho)
        icono = '📱' if ancho < 600 else ('📟' if ancho < 1024 else '🖥️')
        marcos += (f'<div style="flex:none">'
                   f'<div style="font:bold 13px sans-serif;margin-bottom:4px">{icono} {ancho}px</div>'
                   f'<div style="width:{ancho * escala:.0f}px;height:{alto}px;overflow:hidden">'
                   f'<div style="width:{ancho}px;transform:scale({escala});transform-origin:0 0">'
                   f'{_marco(codigo_html, f"{ancho}px", round(alto / escala))}</div></div></div>')
    display(HTML(f'<div style="display:flex;gap:16px;overflow-x:auto;padding:8px 0">{marcos}</div>'))


# 4. Analizador de HTML: radiografía semántica y revisor automático
REGIONES = {
    'header': 'banner (encabezado)',
    'nav': 'navigation (navegación)',
    'main': 'main (contenido principal)',
    'section': 'region (sección temática)',
    'article': 'article (contenido independiente)',
    'aside': 'complementary (complementario)',
    'footer': 'contentinfo (pie de página)',
    'form': 'form (formulario)',
}


class _Analizador(HTMLParser):
    def __init__(self):
        super().__init__()
        self.etiquetas = []
        self.doctype = ''
        self.titulo = ''
        self.niveles = []
        self.labels_for = set()
        self.controles = []
        self.botones_vacios = 0
        self.mapa = []
        self._prof = 0
        self._en_title = False
        self._labels_abiertos = 0
        self._boton = None
        self._encabezado = None

    def handle_decl(self, decl):
        self.doctype = decl

    def handle_starttag(self, tag, attrs):
        a = {k: (v or '') for k, v in attrs}
        self.etiquetas.append((tag, a))
        if tag == 'title':
            self._en_title = True
        if tag == 'label':
            self._labels_abiertos += 1
            if a.get('for'):
                self.labels_for.add(a['for'])
        if tag in ('input', 'select', 'textarea'):
            self.controles.append((tag, a, self._labels_abiertos > 0))
        if tag == 'button':
            self._boton = 'x' if a.get('aria-label') else ''
        if tag in REGIONES:
            self.mapa.append((self._prof, f"🏷️  <{tag}>  →  {REGIONES[tag]}"))
            self._prof += 1
        if re.fullmatch(r'h[1-6]', tag):
            self._encabezado = [int(tag[1]), '']
            self.niveles.append(int(tag[1]))

    def handle_endtag(self, tag):
        if tag == 'title':
            self._en_title = False
        if tag == 'label':
            self._labels_abiertos = max(0, self._labels_abiertos - 1)
        if tag == 'button' and self._boton is not None:
            if not self._boton.strip():
                self.botones_vacios += 1
            self._boton = None
        if tag in REGIONES:
            self._prof = max(0, self._prof - 1)
        if self._encabezado and tag == f'h{self._encabezado[0]}':
            nivel, texto = self._encabezado
            self.mapa.append((self._prof, f"📑 h{nivel}: {' '.join(texto.split())}"))
            self._encabezado = None

    def handle_data(self, data):
        if self._en_title:
            self.titulo += data
        if self._boton is not None:
            self._boton += data
        if self._encabezado:
            self._encabezado[1] += data


def _analizar(codigo_html):
    analizador = _Analizador()
    analizador.feed(codigo_html)
    return analizador


def radiografia(codigo_html, nombre='la página'):
    a = _analizar(codigo_html)
    print(f"🔎 Radiografía de {nombre}: lo que entiende un navegador, Google o un lector de pantalla")
    print("-" * 75)
    if not a.mapa:
        print("⚠️  No encontré regiones ni encabezados.")
        print("    Para un lector de pantalla, esta página es un solo bloque de texto sin estructura.")
    for profundidad, linea in a.mapa:
        print("    " * profundidad + linea)
    etiquetas = [t for t, _ in a.etiquetas]
    semanticas = sum(etiquetas.count(t) for t in REGIONES)
    print("-" * 75)
    print(f"📊 Etiquetas semánticas: {semanticas}   |   <div>: {etiquetas.count('div')}")


def revisar_html(codigo_html, nombre='la maqueta'):
    a = _analizar(codigo_html)
    etiquetas = [t for t, _ in a.etiquetas]
    de_tipo = lambda tag: [attrs for t, attrs in a.etiquetas if t == tag]
    metas = de_tipo('meta')
    raiz = de_tipo('html')
    semanticas = sum(etiquetas.count(t) for t in REGIONES)

    sin_label = []
    for tag, attrs, dentro_de_label in a.controles:
        if attrs.get('type') in ('hidden', 'submit', 'button', 'reset', 'image'):
            continue
        tiene = (dentro_de_label or attrs.get('id') in a.labels_for
                 or attrs.get('aria-label') or attrs.get('aria-labelledby'))
        if not tiene:
            sin_label.append(attrs.get('id') or attrs.get('name') or attrs.get('placeholder') or tag)

    botones_falsos = [x.get('class') for t, x in a.etiquetas
                      if t in ('div', 'span') and re.search(r'bot[oó]n|btn', x.get('class', ''))]
    saltos = [f"h{p}→h{n}" for p, n in zip(a.niveles, a.niveles[1:]) if n > p + 1]
    imgs_sin_alt = [x.get('src', '?') for x in de_tipo('img') if 'alt' not in x]

    pruebas = [
        (a.doctype.lower().startswith('doctype html'),
         'Declara <!DOCTYPE html>', 'Escribe <!DOCTYPE html> en la primera línea.'),
        (bool(raiz) and bool(raiz[0].get('lang')),
         'La etiqueta <html> indica el idioma', 'Usa <html lang="es"> para que el lector de pantalla pronuncie en español.'),
        (any('charset' in m for m in metas),
         'Declara la codificación (meta charset)', 'Agrega <meta charset="UTF-8"> para que se vean bien los acentos y la ñ.'),
        (any(m.get('name') == 'viewport' for m in metas),
         'Tiene meta viewport (lista para celular)', 'Agrega <meta name="viewport" content="width=device-width, initial-scale=1.0">.'),
        (bool(a.titulo.strip()),
         'Tiene un <title> descriptivo', 'Escribe un <title> con el nombre de tu app.'),
        (etiquetas.count('h1') == 1,
         'Usa exactamente un <h1>', f'Hay {etiquetas.count("h1")} <h1>; deja solo uno, con el nombre de la app o de la pantalla.'),
        (bool(a.niveles) and not saltos,
         'Los encabezados siguen un orden (sin saltos)', 'Ordena los títulos h1 → h2 → h3. ' + (f'Saltos: {", ".join(saltos)}' if saltos else 'No hay encabezados.')),
        ('header' in etiquetas,
         'Tiene <header>', 'Envuelve el logo y el título de tu app en <header>.'),
        ('nav' in etiquetas,
         'Tiene <nav>', 'Pon el menú principal dentro de <nav>.'),
        (etiquetas.count('main') == 1,
         'Tiene un solo <main>', f'Hay {etiquetas.count("main")} <main>; el contenido principal va en un único <main>.'),
        ('footer' in etiquetas,
         'Tiene <footer>', 'Agrega un <footer> con los créditos de tu proyecto.'),
        (not imgs_sin_alt,
         'Todas las imágenes tienen texto alternativo (alt)', f'Imágenes sin alt: {", ".join(imgs_sin_alt)}'),
        (not sin_label,
         'Cada campo del formulario tiene <label>', f'Campos sin etiqueta: {", ".join(sin_label)}. El placeholder NO sustituye a <label>.'),
        (a.botones_vacios == 0,
         'Todos los botones tienen texto', 'Escribe un texto dentro de cada <button>, o agrega aria-label.'),
        (not botones_falsos,
         'Los botones son <button> (no <div> disfrazados)', 'Un <div class="boton"> no se puede activar con teclado ni lo anuncia el lector de pantalla. Usa <button>.'),
        ('TODO' not in codigo_html,
         'No quedan TODO pendientes', f'Quedan {codigo_html.count("TODO")} TODO por completar con los datos de tu proyecto.'),
        ('table' not in etiquetas or 'th' in etiquetas,
         'Las tablas tienen encabezados <th>', 'Usa <th scope="col"> en la primera fila de tus tablas.'),
        (etiquetas.count('div') <= semanticas,
         'Sin "divitis" (más etiquetas semánticas que <div>)', f'Hay {etiquetas.count("div")} <div> y {semanticas} etiquetas semánticas. ¿Cuáles <div> podrían ser header, nav, section…?'),
    ]

    aprobadas = sum(ok for ok, _, _ in pruebas)
    print(f"🧪 Revisión automática de {nombre}")
    print("=" * 75)
    for ok, criterio, consejo in pruebas:
        print(f"✅ {criterio}" if ok else f"❌ {criterio}\n      💡 {consejo}")
    print("=" * 75)
    porcentaje = round(100 * aprobadas / len(pruebas))
    medalla = '🏆' if porcentaje == 100 else ('👍' if porcentaje >= 80 else '🔧')
    print(f"{medalla} Resultado: {aprobadas}/{len(pruebas)} criterios ({porcentaje} %)")
    return porcentaje


# 5. Predicciones: ¿qué imprimirá este código?
def predecir(codigo, mi_prediccion):
    if not str(mi_prediccion).strip():
        print("✏️ Primero escribe tu predicción en MI_PREDICCION y vuelve a ejecutar.")
        return
    salida = io.StringIO()
    with contextlib.redirect_stdout(salida):
        exec(codigo, {})
    real = salida.getvalue().strip()
    if ' '.join(str(mi_prediccion).split()) == ' '.join(real.split()):
        print(f"✅ ¡Acertaste! El código imprime: {real}")
    else:
        print(f"🤔 Tu predicción:  {mi_prediccion}")
        print(f"   Resultado real: {real}")
        print("   💬 Coméntalo con tu pareja: ¿en qué línea cambió lo que esperabas?")


# 6. Autoverificador: cada prueba es (descripción, función que devuelve True/False, pista)
def autoverificar(titulo, pruebas, meta=None):
    aprobadas = 0
    print(f"✅ Autoverificador · {titulo}")
    print("=" * 75)
    for descripcion, prueba, pista in pruebas:
        try:
            ok = bool(prueba())
            error = ''
        except Exception as e:
            ok = False
            error = f" ({type(e).__name__}: {e})"
        aprobadas += ok
        print(f"✅ {descripcion}" if ok else f"❌ {descripcion}{error}\n      💡 {pista}")
    print("=" * 75)
    porcentaje = round(100 * aprobadas / len(pruebas))
    meta_txt = f" · meta: {meta}" if meta else ''
    medalla = '🏆' if porcentaje == 100 else ('👍' if porcentaje >= 75 else '🔧')
    print(f"{medalla} Resultado: {aprobadas}/{len(pruebas)} ({porcentaje} %){meta_txt}")
    return porcentaje


# 7. Guardar y leer el trabajo del proyecto (JSON en la carpeta docs/)
def guardar_json(nombre, datos):
    ruta = os.path.join(RUTA_PROYECTO, 'docs', nombre)
    os.makedirs(os.path.dirname(ruta), exist_ok=True)
    with open(ruta, 'w', encoding='utf-8') as f:
        json.dump(datos, f, ensure_ascii=False, indent=2)
    print(f"💾 Guardado en tu proyecto: docs/{nombre}")


def leer_json(nombre, por_defecto=None):
    ruta = os.path.join(RUTA_PROYECTO, 'docs', nombre)
    if not os.path.exists(ruta):
        return por_defecto
    with open(ruta, encoding='utf-8') as f:
        return json.load(f)


# 8. Bitácoras: tickets de salida y uso de IA
def _anotar(archivo, encabezado, fila):
    ruta = os.path.join(RUTA_PROYECTO, 'bitacora', archivo)
    os.makedirs(os.path.dirname(ruta), exist_ok=True)
    nuevo = not os.path.exists(ruta)
    with open(ruta, 'a', encoding='utf-8', newline='') as f:
        escritor = csv.writer(f)
        if nuevo:
            escritor.writerow(encabezado)
        escritor.writerow(fila)


def ticket_salida(sesion, aprendi, me_bloqueo):
    if not aprendi.strip():
        print("✏️ Escribe qué aprendiste hoy antes de registrar tu ticket.")
        return
    _anotar('tickets_salida.csv', ['fecha', 'sesion', 'aprendi', 'me_bloqueo'],
            [f"{datetime.now():%Y-%m-%d %H:%M}", sesion, aprendi.strip(), me_bloqueo.strip()])
    print(f"🎟️ Ticket de la sesión {sesion} registrado en bitacora/tickets_salida.csv")


def registrar_ia(pregunta, respuesta, que_acepte, como_verifique):
    _anotar('bitacora_ia.csv', ['fecha', 'pregunta', 'respuesta', 'que_acepte', 'como_verifique'],
            [f"{datetime.now():%Y-%m-%d %H:%M}", pregunta, respuesta, que_acepte, como_verifique])
    print("🤖 Uso de IA registrado en bitacora/bitacora_ia.csv")


# 9. Ver el árbol de carpetas del proyecto
def arbol(ruta=None, prefijo=''):
    ruta = ruta or RUTA_PROYECTO
    if not prefijo:
        print(f"📁 {os.path.basename(ruta)}/")
    elementos = sorted(os.listdir(ruta))
    for i, nombre in enumerate(elementos):
        ultimo = i == len(elementos) - 1
        completo = os.path.join(ruta, nombre)
        print(f"{prefijo}{'└── ' if ultimo else '├── '}{'📁 ' if os.path.isdir(completo) else '📄 '}{nombre}")
        if os.path.isdir(completo):
            arbol(completo, prefijo + ('    ' if ultimo else '│   '))
# 10. Datos de la semana 3 y calificación automática
ALUMNO = ""  #@param {type:"string"}
GRUPO = "5G"  #@param ["5G", "5H", "5I"]

import hashlib

_ficha = leer_json('ficha.json', {})
NOMBRE_ERP = _ficha.get('nombre_erp') or 'Mi ERP'
INTEGRANTES = _ficha.get('integrantes') or ['Integrante 1', 'Integrante 2']

# ----------------------------------------------------------------------------------------------
# Tabla de puntos de la semana (suma 10). Cada actividad registra su MEJOR intento.
# ----------------------------------------------------------------------------------------------
_ITEMS = [
    # (clave, bloque, actividad, máximo)
    ('p1', 'Participación', 'Sesión 1 · Predicción', 0.10),
    ('t1', 'Participación', 'Sesión 1 · Ticket de salida', 0.15),
    ('p2', 'Participación', 'Sesión 2 · Predicción', 0.10),
    ('t2', 'Participación', 'Sesión 2 · Ticket de salida', 0.15),
    ('p3', 'Participación', 'Sesión 3 · Predicción', 0.10),
    ('t3', 'Participación', 'Sesión 3 · Ticket de salida', 0.15),
    ('p4', 'Participación', 'Sesión 4 · Predicción', 0.10),
    ('t4', 'Participación', 'Sesión 4 · Ticket de salida', 0.15),
    ('e1a', 'Entregable 1 · Estructura semántica', 'Ejercicio guiado: de cajas a significado', 1.00),
    ('e1b', 'Entregable 1 · Estructura semántica', 'Reto ERP: index.html (tablero)', 2.00),
    ('e2a', 'Entregable 2 · Tabla accesible', 'Ejercicio guiado: arregla la tabla', 0.75),
    ('e2b', 'Entregable 2 · Tabla accesible', 'Reto ERP: inventario.html', 1.75),
    ('e3a', 'Entregable 3 · Formulario, maqueta y revisión', 'Ejercicio guiado: arregla el formulario', 0.75),
    ('e3b', 'Entregable 3 · Formulario, maqueta y revisión', 'Reto ERP: formularios.html', 1.00),
    ('e3c', 'Entregable 3 · Formulario, maqueta y revisión', 'Maqueta de 4 páginas enlazadas (15/18 o más)', 1.00),
    ('e3d', 'Entregable 3 · Formulario, maqueta y revisión', 'Revisión de código entre parejas', 0.75),
]
_MAX = {c: m for c, _, _, m in _ITEMS}
_NOMBRE_CALIF = 'calificacion_s3.json'
_CALIF = {}   # clave -> fracción lograda (0 a 1), mejor intento


def _total_posible():
    return round(sum(_MAX.values()), 2)


def _puntos(claves=None):
    return sum(_MAX[c] * _CALIF.get(c, 0) for c in (claves or _MAX))


def _guardar_calif():
    ruta = os.path.join(RUTA_PROYECTO, 'docs', _NOMBRE_CALIF)
    os.makedirs(os.path.dirname(ruta), exist_ok=True)
    with open(ruta, 'w', encoding='utf-8') as f:
        json.dump({'fracciones': _CALIF, 'actualizado': f"{datetime.now():%Y-%m-%d %H:%M}"}, f, indent=1)


def _cargar_calif():
    datos = leer_json(_NOMBRE_CALIF, {}) or {}
    for clave, fraccion in (datos.get('fracciones') or {}).items():
        if clave in _MAX:
            _CALIF[clave] = max(0.0, min(1.0, float(fraccion)))
    if _CALIF:
        print(f"📈 Recuperé tu avance anterior: {_puntos():.2f} / {_total_posible():g} puntos.")


def calif_registrar(clave, fraccion):
    """Guarda tu mejor intento de una actividad y muestra tu calificación acumulada."""
    if clave not in _MAX:
        return
    fraccion = max(0.0, min(1.0, float(fraccion)))
    _CALIF[clave] = max(_CALIF.get(clave, 0.0), round(fraccion, 4))
    _guardar_calif()
    actividad = next(a for c, _, a, _ in _ITEMS if c == clave)
    print(f"🧮 {actividad}: {_MAX[clave] * fraccion:.2f} de {_MAX[clave]:.2f} pts"
          f"  (tu mejor intento cuenta: {_MAX[clave] * _CALIF[clave]:.2f})")
    print(f"📈 CALIFICACIÓN ACUMULADA: {_puntos():.2f} / {_total_posible():g}")


# Las herramientas de la semana 1 ahora también registran puntos (si les pasas la clave)
_predecir_base = predecir
_autoverificar_base = autoverificar
_ticket_base = ticket_salida


def predecir(codigo, mi_prediccion, clave=None):
    _predecir_base(codigo, mi_prediccion)
    if clave and str(mi_prediccion).strip():
        calif_registrar(clave, 1.0)


def autoverificar(titulo, pruebas, meta=None, clave=None):
    porcentaje = _autoverificar_base(titulo, pruebas, meta)
    if clave:
        # Regla de puntos: se otorgan cuando el archivo existe y ya no queda ningún TODO
        bloqueo = []
        for descripcion, prueba, _ in pruebas:
            if descripcion.startswith(('No quedan', 'El archivo', 'Existen')):
                try:
                    ok = bool(prueba())
                except Exception:
                    ok = False
                if not ok:
                    bloqueo.append(descripcion)
        if bloqueo:
            print("🔒 Aún no sumas puntos: primero completa «" + bloqueo[0] + "» (guarda el archivo y quita todos los TODO).")
            calif_registrar(clave, 0)
        else:
            calif_registrar(clave, porcentaje / 100)
    return porcentaje


def ticket_salida(sesion, aprendi, me_bloqueo):
    if len(aprendi.strip()) < 15:
        print("✏️ Escribe con tus palabras qué aprendiste hoy (al menos una frase) y vuelve a ejecutar.")
        return
    _ticket_base(sesion, aprendi, me_bloqueo)
    calif_registrar('t' + str(sesion).split('.')[-1], 1.0)


def calificacion():
    """Tabla con tu calificación acumulada."""
    filas, bloque = '', None
    for clave, grupo, actividad, maximo in _ITEMS:
        if grupo != bloque:
            bloque = grupo
            pts_b = _puntos([c for c, g, _, _ in _ITEMS if g == grupo])
            max_b = sum(m for c, g, _, m in _ITEMS if g == grupo)
            filas += (f'<tr><td colspan="3" style="background:#edf2f7;font-weight:bold;padding:6px 10px">{grupo}'
                      f'</td><td style="background:#edf2f7;font-weight:bold;text-align:right;padding:6px 10px">'
                      f'{pts_b:.2f} / {max_b:.2f}</td></tr>')
        f = _CALIF.get(clave, 0)
        icono = '✅' if f >= 0.999 else ('🔶' if f > 0 else '⬜')
        barra = (f'<div style="background:#e2e8f0;border-radius:4px;height:10px;width:110px">'
                 f'<div style="background:{"#2f855a" if f >= 0.999 else "#b7791f"};height:10px;border-radius:4px;'
                 f'width:{int(110 * f)}px"></div></div>')
        filas += (f'<tr><td style="padding:4px 10px">{icono}</td><td style="padding:4px 10px">{actividad}</td>'
                  f'<td style="padding:4px 10px">{barra}</td>'
                  f'<td style="padding:4px 10px;text-align:right">{_MAX[clave] * f:.2f} / {_MAX[clave]:.2f}</td></tr>')
    total = _puntos()
    display(HTML(
        '<div style="font-family:system-ui,sans-serif;color:#1a202c;max-width:760px">'
        '<table style="border-collapse:collapse;width:100%;font-size:14px">'
        '<caption style="text-align:left;font-weight:bold;font-size:16px;padding:6px 0">📈 Mi calificación acumulada · Semana 3</caption>'
        f'<thead><tr><th scope="col"></th><th scope="col" style="text-align:left;padding:4px 10px">Actividad</th>'
        f'<th scope="col" style="text-align:left;padding:4px 10px">Avance</th>'
        f'<th scope="col" style="text-align:right;padding:4px 10px">Puntos</th></tr></thead>'
        f'<tbody>{filas}</tbody>'
        f'<tfoot><tr><th scope="row" colspan="3" style="text-align:right;padding:8px 10px;font-size:16px">TOTAL</th>'
        f'<td style="text-align:right;padding:8px 10px;font-size:18px;font-weight:bold;color:#1a365d">'
        f'{total:.2f} / {_total_posible():g}</td></tr></tfoot></table></div>'))
    return round(total, 2)


# ----------------------------------------------------------------------------------------------
# Boleta de calificación (se queda guardada en el cuaderno que entregas por Classroom)
# ----------------------------------------------------------------------------------------------
_SAL = 'cetis125-erp-s3-2026'
_ARCHIVOS_ENTREGA = ['index.html', 'inventario.html', 'formularios.html', 'ventas.html', 'clientes.html']


def _datos_boleta():
    return '|'.join(f"{c}:{_CALIF.get(c, 0):.4f}" for c in _MAX)


def _codigo_boleta(alumno, grupo, datos):
    base = f"{_SAL}|{alumno.strip().lower()}|{grupo}|{datos}"
    return hashlib.sha256(base.encode('utf-8')).hexdigest()[:10].upper()


def boleta():
    """Genera tu boleta final. Ejecútala al terminar y guarda el cuaderno con Ctrl+S."""
    if len(ALUMNO.strip()) < 5:
        print("⚠️ Escribe tu nombre completo en el campo ALUMNO de la CELDA 0, ejecútala de nuevo y regresa aquí.")
        return
    total = round(_puntos(), 2)
    datos = _datos_boleta()
    codigo = _codigo_boleta(ALUMNO, GRUPO, datos)
    ahora = f"{datetime.now():%d/%m/%Y %H:%M}"
    entregas = ''
    for nombre in _ARCHIVOS_ENTREGA:
        ruta = os.path.join(RUTA_PROYECTO, nombre)
        if os.path.exists(ruta):
            texto = open(ruta, encoding='utf-8').read()
            estado = 'con TODO pendientes' if 'TODO' in texto else 'completo'
            entregas += f'<li>📄 {nombre} · {len(texto)} caracteres · {estado}</li>'
        else:
            entregas += f'<li>⬜ {nombre} · no encontrado</li>'
    veredicto = ('🏆 Excelente' if total >= 9 else '👍 Aprobado' if total >= 6 else '🔧 Te falta avanzar: repite los autoverificadores')
    filas = ''
    for clave, grupo, actividad, maximo in _ITEMS:
        filas += (f'<tr><td style="padding:3px 8px">{grupo.split(" · ")[0]}</td><td style="padding:3px 8px">{actividad}</td>'
                  f'<td style="padding:3px 8px;text-align:right">{maximo * _CALIF.get(clave, 0):.2f} / {maximo:.2f}</td></tr>')
    html = (
        '<div style="font-family:system-ui,sans-serif;color:#1a202c;max-width:760px;border:3px solid #1a365d;'
        'border-radius:12px;padding:18px;background:#fff">'
        '<h2 style="margin:0 0 4px;color:#1a365d">📊 Boleta de calificación · Semana 3</h2>'
        '<p style="margin:0 0 10px">CETIS 125 · Submódulo 1: Construye aplicaciones web · HTML5 semántico, tablas y formularios</p>'
        f'<p style="margin:2px 0"><b>Alumno:</b> {html_lib.escape(ALUMNO.strip())} · <b>Grupo:</b> {GRUPO}</p>'
        f'<p style="margin:2px 0"><b>ERP:</b> {html_lib.escape(NOMBRE_ERP)} · <b>Fecha:</b> {ahora}</p>'
        f'<table style="border-collapse:collapse;width:100%;font-size:13px;margin:10px 0">'
        '<caption style="text-align:left;font-weight:bold">Detalle de puntos</caption>'
        '<thead><tr><th scope="col" style="text-align:left;padding:3px 8px">Bloque</th>'
        '<th scope="col" style="text-align:left;padding:3px 8px">Actividad</th>'
        '<th scope="col" style="text-align:right;padding:3px 8px">Puntos</th></tr></thead>'
        f'<tbody>{filas}</tbody></table>'
        f'<p style="font-size:24px;margin:8px 0"><b>Calificación formativa: {total:.2f} / 10</b> · {veredicto}</p>'
        f'<details open><summary><b>Archivos de tu proyecto</b></summary><ul style="margin:4px 0">{entregas}</ul></details>'
        f'<p style="font-size:12px;color:#4a5568;margin:10px 0 0">Código de verificación: <b>{codigo}</b><br>'
        f'<span style="word-break:break-all">Datos: {datos}</span></p></div>')
    display(HTML(html))
    with open(os.path.join(RUTA_PROYECTO, 'docs', 'boleta_s3.html'), 'w', encoding='utf-8') as f:
        f.write(html)
    print("💾 Copia guardada en tu proyecto: docs/boleta_s3.html")
    print("📌 AHORA: guarda el cuaderno con Ctrl+S (o Archivo → Guardar) y compártelo por Classroom.")
    return total


def verificar_boleta(alumno, grupo, datos, codigo):
    """(Docente) Comprueba que una boleta no fue alterada y recalcula la calificación."""
    esperado = _codigo_boleta(alumno, grupo, datos)
    total = 0.0
    for par in datos.split('|'):
        c, f = par.split(':')
        total += _MAX.get(c, 0) * float(f)
    if esperado == codigo.strip().upper():
        print(f"✅ Boleta auténtica. Calificación recalculada: {total:.2f} / {_total_posible():g}")
    else:
        print(f"❌ El código no coincide (esperado {esperado}). La boleta pudo ser modificada.")
    return round(total, 2)


# ----------------------------------------------------------------------------------------------
# Herramientas de análisis de HTML (semana 3)
# ----------------------------------------------------------------------------------------------
def leer_pagina(nombre):
    ruta = os.path.join(RUTA_PROYECTO, nombre)
    if not os.path.exists(ruta):
        return ''
    with open(ruta, encoding='utf-8') as f:
        return f.read()


def contar(codigo_html, etiqueta):
    return [t for t, _ in _analizar(codigo_html).etiquetas].count(etiqueta)


def semanticas(codigo_html):
    etiquetas = [t for t, _ in _analizar(codigo_html).etiquetas]
    return sum(etiquetas.count(t) for t in REGIONES)


def texto_de(codigo_html, etiqueta):
    m = re.search(rf'<{etiqueta}\b[^>]*>(.*?)</{etiqueta}>', codigo_html, re.S | re.I)
    return ' '.join(re.sub(r'<[^>]+>', ' ', m.group(1)).split()) if m else ''


def enlaces(codigo_html):
    return [a.get('href', '') for t, a in _analizar(codigo_html).etiquetas if t == 'a']


def criterios_html(codigo_html):
    """Devuelve (aprobados, total) de revisar_html sin imprimir nada."""
    with contextlib.redirect_stdout(io.StringIO()) as buf:
        revisar_html(codigo_html)
    m = re.search(r'Resultado: (\d+)/(\d+)', buf.getvalue())
    return (int(m.group(1)), int(m.group(2))) if m else (0, 18)


def sin_divitis(codigo_html):
    return semanticas(codigo_html) >= 1 and contar(codigo_html, 'div') <= semanticas(codigo_html)


class _Tablas(HTMLParser):
    def __init__(self):
        super().__init__()
        self.tablas, self._t, self._fila, self._sec, self._cap = [], None, None, None, False

    def handle_starttag(self, tag, attrs):
        a = {k: (v or '') for k, v in attrs}
        if tag == 'table':
            self._t = {'caption': '', 'thead': False, 'tbody': False, 'tfoot': False, 'th': [], 'filas': [],
                       'maquillaje': [k for k in a if k in ('border', 'bgcolor', 'align', 'cellpadding', 'cellspacing', 'width')]}
            self.tablas.append(self._t)
        elif self._t is None:
            return
        elif tag == 'caption':
            self._cap = True
        elif tag in ('thead', 'tbody', 'tfoot'):
            self._t[tag] = True
            self._sec = tag
        elif tag == 'tr':
            self._fila = {'sec': self._sec, 'celdas': 0, 'th': 0, 'td': 0}
            self._t['filas'].append(self._fila)
        elif tag in ('th', 'td') and self._fila is not None:
            self._fila['celdas'] += 1
            self._fila[tag] += 1
            if tag == 'th':
                self._t['th'].append({'scope': a.get('scope', ''), 'sec': self._sec})

    def handle_endtag(self, tag):
        if tag == 'caption':
            self._cap = False
        elif tag in ('thead', 'tbody', 'tfoot'):
            self._sec = None
        elif tag == 'tr':
            self._fila = None
        elif tag == 'table':
            self._t = None

    def handle_data(self, data):
        if self._cap and self._t is not None:
            self._t['caption'] += data


def tablas(codigo_html):
    p = _Tablas()
    p.feed(codigo_html)
    return p.tablas


def tabla_envuelta(codigo_html):
    """¿La tabla vive dentro de un contenedor con overflow-x: auto?"""
    return bool(re.search(r'overflow-x\s*:\s*(auto|scroll)', codigo_html)) and \
        bool(re.search(r'<(div|section)[^>]*class="[^"]*(scroll|tabla)[^"]*"[^>]*>\s*<table', codigo_html, re.S))


def analizar_tablas(codigo_html, nombre='la página'):
    print(f"📊 Radiografía de las tablas de {nombre}")
    print("-" * 75)
    encontradas = tablas(codigo_html)
    if not encontradas:
        print("⚠️ No encontré ninguna <table>.")
    for i, t in enumerate(encontradas, 1):
        cols = max([f['celdas'] for f in t['filas']] or [0])
        cuerpo = [f for f in t['filas'] if f['sec'] == 'tbody']
        print(f"Tabla {i}: {len(t['filas'])} filas · {cols} columnas")
        print(f"   {'✅' if t['caption'].strip() else '❌'} <caption>: {' '.join(t['caption'].split()) or '(falta)'}")
        print(f"   {'✅' if t['thead'] else '❌'} <thead>    {'✅' if t['tbody'] else '❌'} <tbody>    {'✅' if t['tfoot'] else '➖'} <tfoot>")
        con_scope = sum(1 for h in t['th'] if h['scope'] in ('col', 'row'))
        print(f"   {'✅' if t['th'] and con_scope == len(t['th']) else '❌'} <th> con scope: {con_scope} de {len(t['th'])}")
        if t['maquillaje']:
            print(f"   ⚠️ Atributos viejos de maquillaje: {', '.join(t['maquillaje'])} (el aspecto va en CSS)")
        if '<html' in codigo_html.lower():
            print(f"   📦 Contenedor con scroll horizontal: {'✅' if tabla_envuelta(codigo_html) else '❌'}")
    print("-" * 75)


class _Formularios(HTMLParser):
    def __init__(self):
        super().__init__()
        self.formularios, self._f, self._lab, self._btn = [], None, 0, None

    def handle_starttag(self, tag, attrs):
        a = {k: (v or '') for k, v in attrs}
        if tag == 'form':
            self._f = {'fieldset': 0, 'legend': 0, 'controles': [], 'labels_for': set(), 'botones': []}
            self.formularios.append(self._f)
            return
        if self._f is None:
            return
        if tag in ('fieldset', 'legend'):
            self._f[tag] += 1
        elif tag == 'label':
            self._lab += 1
            if a.get('for'):
                self._f['labels_for'].add(a['for'])
        elif tag in ('input', 'select', 'textarea'):
            self._f['controles'].append({'tag': tag, 'a': a, 'envuelto': self._lab > 0})
        elif tag == 'button':
            self._btn = {'a': a, 'texto': ''}

    def handle_endtag(self, tag):
        if tag == 'label':
            self._lab = max(0, self._lab - 1)
        elif tag == 'button' and self._btn is not None and self._f is not None:
            self._f['botones'].append(self._btn)
            self._btn = None
        elif tag == 'form':
            self._f = None

    def handle_data(self, data):
        if self._btn is not None:
            self._btn['texto'] += data


def formularios(codigo_html):
    p = _Formularios()
    p.feed(codigo_html)
    for f in p.formularios:
        f['campos'] = [c for c in f['controles'] if c['a'].get('type') not in ('hidden', 'submit', 'button', 'reset', 'image')]
        for c in f['campos']:
            c['etiqueta'] = bool(c['envuelto'] or c['a'].get('id') in f['labels_for'] or c['a'].get('aria-label'))
    return p.formularios


def analizar_formularios(codigo_html, nombre='la página'):
    print(f"📝 Radiografía de los formularios de {nombre}")
    print("-" * 75)
    encontrados = formularios(codigo_html)
    if not encontrados:
        print("⚠️ No encontré ningún <form>.")
    for i, f in enumerate(encontrados, 1):
        print(f"Formulario {i}: {len(f['campos'])} campos · fieldset: {f['fieldset']} · legend: {f['legend']} · botones: {len(f['botones'])}")
        for c in f['campos']:
            a = c['a']
            tipo = a.get('type', 'text') if c['tag'] == 'input' else c['tag']
            reglas = ', '.join(r for r in ('required', 'min', 'max', 'minlength', 'maxlength', 'pattern') if r in a) or '—'
            print(f"   {'✅' if c['etiqueta'] else '❌'} {a.get('name') or a.get('id') or '(sin nombre)':<14} tipo: {tipo:<9} label: {'sí' if c['etiqueta'] else 'NO'}   reglas: {reglas}")
    print("-" * 75)


# ----------------------------------------------------------------------------------------------
# Menú común del ERP (consistencia: Nielsen #4)
# ----------------------------------------------------------------------------------------------
PAGINAS_ERP = [('index.html', 'Tablero'), ('inventario.html', 'Inventario'), ('ventas.html', 'Ventas'), ('clientes.html', 'Clientes')]


def menu_erp(activa):
    items = ''
    for archivo, texto in PAGINAS_ERP:
        actual = ' aria-current="page"' if archivo == activa else ''
        items += f'<li><a href="{archivo}"{actual}>{texto}</a></li>\n    '
    return f'<nav aria-label="Principal">\n  <ul>\n    {items.rstrip()}\n  </ul>\n</nav>'


def poner_menu(archivo):
    """Reemplaza el <nav> de un archivo por el menú común del ERP."""
    codigo = leer_pagina(archivo)
    if not re.search(r'<nav\b.*?</nav>', codigo, re.S):
        print(f"⚠️ {archivo} no tiene <nav>; agrégalo con el menú: {menu_erp(archivo)}")
        return
    nuevo = re.sub(r'<nav\b.*?</nav>', lambda m: menu_erp(archivo), codigo, count=1, flags=re.S)
    guardar(archivo, nuevo)


# Revisión entre parejas
def revision_pares(codigo_otra_pareja, nombre_pareja, fortaleza_1, fortaleza_2, deseo):
    codigo = codigo_otra_pareja.strip()
    if len(codigo) < 100 or '<' not in codigo:
        print("✏️ Pega en CODIGO_OTRA_PAREJA el HTML de la pareja que revisas y vuelve a ejecutar.")
    else:
        revisar_html(codigo, f"la maqueta de {nombre_pareja or 'la otra pareja'}")
        print()
    propias = [leer_pagina(n).strip() for n in _ARCHIVOS_ENTREGA]
    concretos = ('header', 'nav', 'main', 'section', 'footer', 'label', 'alt', 'caption', 'scope', 'button', 'required', 'h1', 'div', 'lang', 'viewport', 'tabla', 'formulario', 'menú', 'título')
    pruebas = [
        ("Pegaste el HTML completo de otra pareja (más de 400 caracteres)", lambda: len(codigo) >= 400 and '<html' in codigo.lower(),
         "Copia el código de su página (Ver código fuente) y pégalo entre las comillas triples."),
        ("Ese código NO es el de tu propia maqueta", lambda: codigo not in propias,
         "Revisa la maqueta de OTRA pareja, no la tuya."),
        ("Escribiste el nombre de la pareja revisada", lambda: len(nombre_pareja.strip()) >= 3, "Anota los nombres de la pareja."),
        ("⭐ Fortaleza 1: una frase concreta (20 caracteres o más)", lambda: len(fortaleza_1.strip()) >= 20, "Di qué hicieron bien y dónde."),
        ("⭐ Fortaleza 2: una frase concreta (20 caracteres o más)", lambda: len(fortaleza_2.strip()) >= 20, "Busca una segunda cosa bien hecha."),
        ("💭 Deseo de mejora: una frase concreta (20 caracteres o más)", lambda: len(deseo.strip()) >= 20, "Propón una mejora que puedan hacer hoy."),
        ("Tus tres comentarios son diferentes entre sí", lambda: len({fortaleza_1.strip().lower(), fortaleza_2.strip().lower(), deseo.strip().lower()}) == 3, "No repitas el mismo comentario."),
        ("Tu comentario menciona algo concreto del código (etiqueta, atributo o criterio)",
         lambda: any(p in (fortaleza_1 + fortaleza_2 + deseo).lower() for p in concretos), "Nombra la etiqueta o el criterio: nav, label, h1, alt, caption…"),
    ]
    porcentaje = autoverificar("Revisión de código entre parejas (⭐⭐ + 💭)", pruebas, clave='e3d')
    if porcentaje == 100:
        _anotar('revisiones_pares.csv', ['fecha', 'pareja_revisada', 'fortaleza_1', 'fortaleza_2', 'deseo'],
                [f"{datetime.now():%Y-%m-%d %H:%M}", nombre_pareja, fortaleza_1, fortaleza_2, deseo])
        print("📝 Revisión guardada en bitacora/revisiones_pares.csv")
    return porcentaje


_cargar_calif()


print("=" * 75)
print(f"📁 Carpeta del proyecto:\n{RUTA_PROYECTO}")
print(f"🏪 ERP: {NOMBRE_ERP}" + ("" if ALUMNO.strip() else "   ⚠️ Escribe tu nombre en ALUMNO y ejecuta esta celda otra vez"))
print("🧰 Herramientas de siempre: predecir, autoverificar, guardar, leer, mostrar, lado_a_lado,")
print("   comparar_pantallas, radiografia, revisar_html, ticket_salida, registrar_ia, arbol")
print("   NUEVAS (semana 3): analizar_tablas, analizar_formularios, menu_erp, poner_menu,")
print("   revision_pares, calificacion (tu calificación acumulada) y boleta (tu boleta final)")
print("=" * 75)
````

## 🏷️ A · Semántica: ponerle nombre a cada caja
Una etiqueta **semántica** dice *qué es* su contenido, no solo cómo se ve. `<div>` es una caja sin significado: sirve para agrupar cuando ninguna otra etiqueta aplica *(MDN, elementos de HTML)*.

| Etiqueta | Qué es | Regla |
|---|---|---|
| `<header>` | Encabezado de la página (logo, nombre del ERP, título) | Va al principio del `<body>` |
| `<nav>` | Menú de navegación principal | Lleva enlaces `<a href>` |
| `<main>` | Contenido principal | **Solo uno** por página |
| `<section>` | Bloque con un tema propio | Siempre con su título (`<h2>`) |
| `<article>` | Contenido que se entiende solo (una tarjeta, un reporte) | Dentro de un `<section>` o del `<main>` |
| `<aside>` | Contenido complementario (atajos, ayuda) | Fuera del flujo principal |
| `<footer>` | Pie de página (créditos, integrantes) | Al final del `<body>` |
| `<h1>` … `<h6>` | Títulos por niveles | **Un solo `<h1>`**; el orden no se salta: h1 → h2 → h3 |

**¿Por qué importa?** *(WCAG 2.1: 1.3.1 Información y relaciones, 2.4.1 Evitar bloques, 2.4.6 Encabezados y etiquetas)*: una persona con lector de pantalla salta de región en región (*“ir a navegación”, “ir al contenido principal”*) y lee la lista de títulos para decidir dónde entrar. Con puros `<div>` no tiene a dónde saltar. Además, la semántica ayuda a que Google entienda tu página.

> ⚠️ **Divitis:** usar `<div>` para todo. Nuestra regla: **nunca debe haber más `<div>` que etiquetas semánticas** (`revisar_html` lo comprueba).

## 📊 A · Anatomía de una tabla accesible
Usa `<table>` **solo para datos con filas y columnas** (nunca para acomodar la página) *(MDN, tablas HTML)*. Una tabla accesible tiene estas piezas:

| `<caption>` | El **título** de la tabla: “Productos en existencia al 5 de noviembre”. Lo lee el lector de pantalla antes de los datos |
| `<thead>` | Agrupa la fila de **encabezados** |
| `<th scope="col">` | Encabezado de **columna**. `scope` le dice al lector de pantalla qué celdas describe *(WCAG 1.3.1 Información y relaciones)* |
| `<th scope="row">` | Encabezado de **fila** (por ejemplo, el nombre del producto) |
| `<tbody>` | Agrupa las filas de datos |
| `<tfoot>` | Totales (opcional) |
| `<td>` | Una celda de dato |

**Dos errores clásicos**
1. Poner los encabezados como `<td><b>…</b></td>`: se *ven* como encabezados, pero para el lector de pantalla no lo son.
2. Usar atributos viejos de maquillaje (`border`, `bgcolor`, `width`): el aspecto se hace con CSS, no con HTML.

**La tabla y el celular.** Una tabla ancha se sale de una pantalla de 360 px. La solución estándar es envolverla en un contenedor con `overflow-x: auto`: así **se desliza dentro de su caja** y el resto de la página no se mueve *(MDN, diseño receptivo)*.

```html
<div class="tabla-scroll"> <table> … </table> </div>
```
```css
.tabla-scroll { overflow-x: auto; }
```

## 📝 A · Anatomía de un formulario accesible
| Pieza | Para qué sirve |
|---|---|
| `<form>` | Agrupa todo lo que se envía junto |
| `<fieldset>` + `<legend>` | Agrupa campos relacionados y les pone **título** (“Datos del producto”) |
| `<label for="x">` + `<input id="x">` | **Etiqueta** vinculada al campo: se lee en voz alta y al tocarla se activa el campo *(WCAG 1.3.1, 3.3.2 Etiquetas o instrucciones)* |
| `type="number" · "email" · "tel" · "date"` | El navegador muestra el teclado correcto y valida el formato *(MDN, formularios web)* |
| `<select>` + `<option>` | Lista de opciones: **reconocer en lugar de recordar** (Nielsen #6) |
| `<textarea>` | Texto largo (notas) |
| `<button type="submit">` | Botón real con texto visible, que funciona con teclado |

**Validación nativa (sin JavaScript)**

| Atributo | Qué hace | Ejemplo en el ERP |
|---|---|---|
| `required` | El campo no puede ir vacío | Nombre del producto |
| `min` / `max` | Rango de un número o fecha | Cantidad mínima 1 |
| `minlength` / `maxlength` | Largo del texto | Nombre de 3 a 40 caracteres |
| `pattern` | Expresión regular | Código con forma `P001` → `pattern="[A-Z][0-9]{3}"` |
| `step` | Salto permitido | Precio con centavos: `step="0.01"` |

**Tres reglas de oro**
1. **El `placeholder` NO es un `label`:** desaparece al escribir y muchos lectores de pantalla no lo anuncian.
2. Cada `id` es **único en toda la página** (dos formularios no pueden usar `id="cantidad"` a la vez).
3. **La validación del navegador es comodidad, no seguridad:** un atacante puede saltarse el HTML. El servidor siempre valida de nuevo *(OWASP, Input Validation Cheat Sheet)*. Lo veremos cuando programemos el *backend*.

## 🔍 A · Revisión de código entre parejas
Una **revisión de código** es leer el trabajo de otra persona para encontrar errores y aprender ideas. Encuentra problemas que quien escribió ya no ve y **todos aprenden de todos**.

**Formato ⭐⭐ + 💭 (dos fortalezas y un deseo)**

| | Qué escribes | Ejemplo vago ❌ | Ejemplo concreto ✅ |
|---|---|---|---|
| ⭐ | Algo que está bien hecho y **dónde** | “Está bien” | “Su `<nav>` tiene los 4 enlaces y marca la página actual con `aria-current`” |
| ⭐ | Una segunda fortaleza | “Me gustó” | “La tabla de inventario tiene `caption` y `th scope` y se desliza en 360 px” |
| 💭 | Una mejora **que se pueda hacer hoy** | “Mejórenlo” | “Me gustaría que el campo Cantidad tuviera `<label for>` porque `revisar_html` marca ❌” |

**Checklist del revisor** (todo lo contesta `revisar_html`, pero tú agregas el criterio humano)
1. ¿`revisar_html` llega a 15/18? ¿Qué ❌ quedan?
2. ¿Hay un solo `<h1>` y los títulos van en orden?
