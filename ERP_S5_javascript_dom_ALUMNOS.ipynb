````python
#@title ▶️ Ejecuta esta celda SIEMPRE primero (doble clic para ver el código) { display-mode: "form" }
import os
import re
import html as html_lib
from html.parser import HTMLParser
from IPython.display import HTML, display

# 1. Montar Google Drive y ubicar la carpeta del proyecto ERP
import io
import csv
import json
import base64
import contextlib
from datetime import datetime

try:
    from google.colab import drive
    print("🔄 Conectando con Google Drive...")
    drive.mount('/content/drive')
    RUTA_DRIVE = '/content/drive/MyDrive'
except ImportError:
    RUTA_DRIVE = os.path.join(os.getcwd(), 'MyDrive')

FICHA = os.path.join(RUTA_DRIVE, 'ERP_ficha_proyecto.json')


def _ruta_del_proyecto():
    # Si ya creaste tu ficha (sesión 2), trabajamos en la carpeta ERP_<nombre de tu ERP>
    if os.path.exists(FICHA):
        with open(FICHA, encoding='utf-8') as f:
            return json.load(f)['carpeta']
    return os.path.join(RUTA_DRIVE, 'ERP_practicas')


RUTA_PROYECTO = _ruta_del_proyecto()
RUTA_SEMANA = RUTA_PROYECTO          # guardar() y leer() trabajan en la carpeta del proyecto
os.makedirs(RUTA_PROYECTO, exist_ok=True)


# 2. Guardar y leer archivos en Drive
def guardar(nombre_archivo, contenido):
    ruta = os.path.join(RUTA_SEMANA, nombre_archivo)
    with open(ruta, 'w', encoding='utf-8') as f:
        f.write(contenido)
    print(f"💾 Guardado en Drive: {ruta}")


def leer(nombre_archivo):
    ruta = os.path.join(RUTA_SEMANA, nombre_archivo)
    with open(ruta, 'r', encoding='utf-8') as f:
        return f.read()


# 3. Mostrar páginas HTML dentro de Colab (cada una en su propio marco)
def _marco(codigo_html, ancho='100%', alto=500):
    return (f'<iframe srcdoc="{html_lib.escape(codigo_html, quote=True)}" '
            f'style="width:{ancho};height:{alto}px;border:1px solid #cbd5e0;'
            f'border-radius:8px;background:#fff"></iframe>')


def mostrar(codigo_html, alto=500):
    display(HTML(_marco(codigo_html, '100%', alto)))


def lado_a_lado(html_a, html_b, titulo_a='A', titulo_b='B', alto=320):
    columnas = ''
    for titulo, pagina in ((titulo_a, html_a), (titulo_b, html_b)):
        columnas += (f'<div style="flex:1;min-width:280px">'
                     f'<div style="font:bold 14px sans-serif;margin-bottom:4px">{titulo}</div>'
                     f'{_marco(pagina, "100%", alto)}</div>')
    display(HTML(f'<div style="display:flex;gap:16px;flex-wrap:wrap">{columnas}</div>'))


def con_css(codigo_html, css):
    # Inserta una hoja de estilos <style> justo antes de </head>
    return codigo_html.replace('</head>', f'<style>\n{css}\n</style>\n</head>', 1)


def comparar_pantallas(codigo_html, anchos=(360, 768, 1280), alto=560):
    marcos = ''
    for ancho in anchos:
        escala = min(1, 380 / ancho)
        icono = '📱' if ancho < 600 else ('📟' if ancho < 1024 else '🖥️')
        marcos += (f'<div style="flex:none">'
                   f'<div style="font:bold 13px sans-serif;margin-bottom:4px">{icono} {ancho}px</div>'
                   f'<div style="width:{ancho * escala:.0f}px;height:{alto}px;overflow:hidden">'
                   f'<div style="width:{ancho}px;transform:scale({escala});transform-origin:0 0">'
                   f'{_marco(codigo_html, f"{ancho}px", round(alto / escala))}</div></div></div>')
    display(HTML(f'<div style="display:flex;gap:16px;overflow-x:auto;padding:8px 0">{marcos}</div>'))


# 4. Analizador de HTML: radiografía semántica y revisor automático
REGIONES = {
    'header': 'banner (encabezado)',
    'nav': 'navigation (navegación)',
    'main': 'main (contenido principal)',
    'section': 'region (sección temática)',
    'article': 'article (contenido independiente)',
    'aside': 'complementary (complementario)',
    'footer': 'contentinfo (pie de página)',
    'form': 'form (formulario)',
}


class _Analizador(HTMLParser):
    def __init__(self):
        super().__init__()
        self.etiquetas = []
        self.doctype = ''
        self.titulo = ''
        self.niveles = []
        self.labels_for = set()
        self.controles = []
        self.botones_vacios = 0
        self.mapa = []
        self._prof = 0
        self._en_title = False
        self._labels_abiertos = 0
        self._boton = None
        self._encabezado = None

    def handle_decl(self, decl):
        self.doctype = decl

    def handle_starttag(self, tag, attrs):
        a = {k: (v or '') for k, v in attrs}
        self.etiquetas.append((tag, a))
        if tag == 'title':
            self._en_title = True
        if tag == 'label':
            self._labels_abiertos += 1
            if a.get('for'):
                self.labels_for.add(a['for'])
        if tag in ('input', 'select', 'textarea'):
            self.controles.append((tag, a, self._labels_abiertos > 0))
        if tag == 'button':
            self._boton = 'x' if a.get('aria-label') else ''
        if tag in REGIONES:
            self.mapa.append((self._prof, f"🏷️  <{tag}>  →  {REGIONES[tag]}"))
            self._prof += 1
        if re.fullmatch(r'h[1-6]', tag):
            self._encabezado = [int(tag[1]), '']
            self.niveles.append(int(tag[1]))

    def handle_endtag(self, tag):
        if tag == 'title':
            self._en_title = False
        if tag == 'label':
            self._labels_abiertos = max(0, self._labels_abiertos - 1)
        if tag == 'button' and self._boton is not None:
            if not self._boton.strip():
                self.botones_vacios += 1
            self._boton = None
        if tag in REGIONES:
            self._prof = max(0, self._prof - 1)
        if self._encabezado and tag == f'h{self._encabezado[0]}':
            nivel, texto = self._encabezado
            self.mapa.append((self._prof, f"📑 h{nivel}: {' '.join(texto.split())}"))
            self._encabezado = None

    def handle_data(self, data):
        if self._en_title:
            self.titulo += data
        if self._boton is not None:
            self._boton += data
        if self._encabezado:
            self._encabezado[1] += data


def _analizar(codigo_html):
    analizador = _Analizador()
    analizador.feed(codigo_html)
    return analizador


def radiografia(codigo_html, nombre='la página'):
    a = _analizar(codigo_html)
    print(f"🔎 Radiografía de {nombre}: lo que entiende un navegador, Google o un lector de pantalla")
    print("-" * 75)
    if not a.mapa:
        print("⚠️  No encontré regiones ni encabezados.")
        print("    Para un lector de pantalla, esta página es un solo bloque de texto sin estructura.")
    for profundidad, linea in a.mapa:
        print("    " * profundidad + linea)
    etiquetas = [t for t, _ in a.etiquetas]
    semanticas = sum(etiquetas.count(t) for t in REGIONES)
    print("-" * 75)
    print(f"📊 Etiquetas semánticas: {semanticas}   |   <div>: {etiquetas.count('div')}")


def revisar_html(codigo_html, nombre='la maqueta'):
    a = _analizar(codigo_html)
    etiquetas = [t for t, _ in a.etiquetas]
    de_tipo = lambda tag: [attrs for t, attrs in a.etiquetas if t == tag]
    metas = de_tipo('meta')
    raiz = de_tipo('html')
    semanticas = sum(etiquetas.count(t) for t in REGIONES)

    sin_label = []
    for tag, attrs, dentro_de_label in a.controles:
        if attrs.get('type') in ('hidden', 'submit', 'button', 'reset', 'image'):
            continue
        tiene = (dentro_de_label or attrs.get('id') in a.labels_for
                 or attrs.get('aria-label') or attrs.get('aria-labelledby'))
        if not tiene:
            sin_label.append(attrs.get('id') or attrs.get('name') or attrs.get('placeholder') or tag)

    botones_falsos = [x.get('class') for t, x in a.etiquetas
                      if t in ('div', 'span') and re.search(r'bot[oó]n|btn', x.get('class', ''))]
    saltos = [f"h{p}→h{n}" for p, n in zip(a.niveles, a.niveles[1:]) if n > p + 1]
    imgs_sin_alt = [x.get('src', '?') for x in de_tipo('img') if 'alt' not in x]

    pruebas = [
        (a.doctype.lower().startswith('doctype html'),
         'Declara <!DOCTYPE html>', 'Escribe <!DOCTYPE html> en la primera línea.'),
        (bool(raiz) and bool(raiz[0].get('lang')),
         'La etiqueta <html> indica el idioma', 'Usa <html lang="es"> para que el lector de pantalla pronuncie en español.'),
        (any('charset' in m for m in metas),
         'Declara la codificación (meta charset)', 'Agrega <meta charset="UTF-8"> para que se vean bien los acentos y la ñ.'),
        (any(m.get('name') == 'viewport' for m in metas),
         'Tiene meta viewport (lista para celular)', 'Agrega <meta name="viewport" content="width=device-width, initial-scale=1.0">.'),
        (bool(a.titulo.strip()),
         'Tiene un <title> descriptivo', 'Escribe un <title> con el nombre de tu app.'),
        (etiquetas.count('h1') == 1,
         'Usa exactamente un <h1>', f'Hay {etiquetas.count("h1")} <h1>; deja solo uno, con el nombre de la app o de la pantalla.'),
        (bool(a.niveles) and not saltos,
         'Los encabezados siguen un orden (sin saltos)', 'Ordena los títulos h1 → h2 → h3. ' + (f'Saltos: {", ".join(saltos)}' if saltos else 'No hay encabezados.')),
        ('header' in etiquetas,
         'Tiene <header>', 'Envuelve el logo y el título de tu app en <header>.'),
        ('nav' in etiquetas,
         'Tiene <nav>', 'Pon el menú principal dentro de <nav>.'),
        (etiquetas.count('main') == 1,
         'Tiene un solo <main>', f'Hay {etiquetas.count("main")} <main>; el contenido principal va en un único <main>.'),
        ('footer' in etiquetas,
         'Tiene <footer>', 'Agrega un <footer> con los créditos de tu proyecto.'),
        (not imgs_sin_alt,
         'Todas las imágenes tienen texto alternativo (alt)', f'Imágenes sin alt: {", ".join(imgs_sin_alt)}'),
        (not sin_label,
         'Cada campo del formulario tiene <label>', f'Campos sin etiqueta: {", ".join(sin_label)}. El placeholder NO sustituye a <label>.'),
        (a.botones_vacios == 0,
         'Todos los botones tienen texto', 'Escribe un texto dentro de cada <button>, o agrega aria-label.'),
        (not botones_falsos,
         'Los botones son <button> (no <div> disfrazados)', 'Un <div class="boton"> no se puede activar con teclado ni lo anuncia el lector de pantalla. Usa <button>.'),
        ('TODO' not in codigo_html,
         'No quedan TODO pendientes', f'Quedan {codigo_html.count("TODO")} TODO por completar con los datos de tu proyecto.'),
        ('table' not in etiquetas or 'th' in etiquetas,
         'Las tablas tienen encabezados <th>', 'Usa <th scope="col"> en la primera fila de tus tablas.'),
        (etiquetas.count('div') <= semanticas,
         'Sin "divitis" (más etiquetas semánticas que <div>)', f'Hay {etiquetas.count("div")} <div> y {semanticas} etiquetas semánticas. ¿Cuáles <div> podrían ser header, nav, section…?'),
    ]

    aprobadas = sum(ok for ok, _, _ in pruebas)
    print(f"🧪 Revisión automática de {nombre}")
    print("=" * 75)
    for ok, criterio, consejo in pruebas:
        print(f"✅ {criterio}" if ok else f"❌ {criterio}\n      💡 {consejo}")
    print("=" * 75)
    porcentaje = round(100 * aprobadas / len(pruebas))
    medalla = '🏆' if porcentaje == 100 else ('👍' if porcentaje >= 80 else '🔧')
    print(f"{medalla} Resultado: {aprobadas}/{len(pruebas)} criterios ({porcentaje} %)")
    return porcentaje


# 5. Predicciones: ¿qué imprimirá este código?
def predecir(codigo, mi_prediccion):
    if not str(mi_prediccion).strip():
        print("✏️ Primero escribe tu predicción en MI_PREDICCION y vuelve a ejecutar.")
        return
    salida = io.StringIO()
    with contextlib.redirect_stdout(salida):
        exec(codigo, {})
    real = salida.getvalue().strip()
    if ' '.join(str(mi_prediccion).split()) == ' '.join(real.split()):
        print(f"✅ ¡Acertaste! El código imprime: {real}")
    else:
        print(f"🤔 Tu predicción:  {mi_prediccion}")
        print(f"   Resultado real: {real}")
        print("   💬 Coméntalo con tu pareja: ¿en qué línea cambió lo que esperabas?")


# 6. Autoverificador: cada prueba es (descripción, función que devuelve True/False, pista)
def autoverificar(titulo, pruebas, meta=None):
    aprobadas = 0
    print(f"✅ Autoverificador · {titulo}")
    print("=" * 75)
    for descripcion, prueba, pista in pruebas:
        try:
            ok = bool(prueba())
            error = ''
        except Exception as e:
            ok = False
            error = f" ({type(e).__name__}: {e})"
        aprobadas += ok
        print(f"✅ {descripcion}" if ok else f"❌ {descripcion}{error}\n      💡 {pista}")
    print("=" * 75)
    porcentaje = round(100 * aprobadas / len(pruebas))
    meta_txt = f" · meta: {meta}" if meta else ''
    medalla = '🏆' if porcentaje == 100 else ('👍' if porcentaje >= 75 else '🔧')
    print(f"{medalla} Resultado: {aprobadas}/{len(pruebas)} ({porcentaje} %){meta_txt}")
    return porcentaje


# 7. Guardar y leer el trabajo del proyecto (JSON en la carpeta docs/)
def guardar_json(nombre, datos):
    ruta = os.path.join(RUTA_PROYECTO, 'docs', nombre)
    os.makedirs(os.path.dirname(ruta), exist_ok=True)
    with open(ruta, 'w', encoding='utf-8') as f:
        json.dump(datos, f, ensure_ascii=False, indent=2)
    print(f"💾 Guardado en tu proyecto: docs/{nombre}")


def leer_json(nombre, por_defecto=None):
    ruta = os.path.join(RUTA_PROYECTO, 'docs', nombre)
    if not os.path.exists(ruta):
        return por_defecto
    with open(ruta, encoding='utf-8') as f:
        return json.load(f)


# 8. Bitácoras: tickets de salida y uso de IA
def _anotar(archivo, encabezado, fila):
    ruta = os.path.join(RUTA_PROYECTO, 'bitacora', archivo)
    os.makedirs(os.path.dirname(ruta), exist_ok=True)
    nuevo = not os.path.exists(ruta)
    with open(ruta, 'a', encoding='utf-8', newline='') as f:
        escritor = csv.writer(f)
        if nuevo:
            escritor.writerow(encabezado)
        escritor.writerow(fila)


def ticket_salida(sesion, aprendi, me_bloqueo):
    if not aprendi.strip():
        print("✏️ Escribe qué aprendiste hoy antes de registrar tu ticket.")
        return
    _anotar('tickets_salida.csv', ['fecha', 'sesion', 'aprendi', 'me_bloqueo'],
            [f"{datetime.now():%Y-%m-%d %H:%M}", sesion, aprendi.strip(), me_bloqueo.strip()])
    print(f"🎟️ Ticket de la sesión {sesion} registrado en bitacora/tickets_salida.csv")


def registrar_ia(pregunta, respuesta, que_acepte, como_verifique):
    _anotar('bitacora_ia.csv', ['fecha', 'pregunta', 'respuesta', 'que_acepte', 'como_verifique'],
            [f"{datetime.now():%Y-%m-%d %H:%M}", pregunta, respuesta, que_acepte, como_verifique])
    print("🤖 Uso de IA registrado en bitacora/bitacora_ia.csv")


# 9. Ver el árbol de carpetas del proyecto
def arbol(ruta=None, prefijo=''):
    ruta = ruta or RUTA_PROYECTO
    if not prefijo:
        print(f"📁 {os.path.basename(ruta)}/")
    elementos = sorted(os.listdir(ruta))
    for i, nombre in enumerate(elementos):
        ultimo = i == len(elementos) - 1
        completo = os.path.join(ruta, nombre)
        print(f"{prefijo}{'└── ' if ultimo else '├── '}{'📁 ' if os.path.isdir(completo) else '📄 '}{nombre}")
        if os.path.isdir(completo):
            arbol(completo, prefijo + ('    ' if ultimo else '│   '))
# 10. Datos de la semana 5 y calificación automática
ALUMNO = ""  #@param {type:"string"}
GRUPO = "5G"  #@param ["5G", "5H", "5I"]

import hashlib

_ficha = leer_json('ficha.json', {})
NOMBRE_ERP = _ficha.get('nombre_erp') or 'Mi ERP'
INTEGRANTES = _ficha.get('integrantes') or ['Integrante 1', 'Integrante 2']

# ----------------------------------------------------------------------------------------------
# Tabla de puntos de la semana (suma 10). Cada actividad registra su MEJOR intento.
# ----------------------------------------------------------------------------------------------
_ITEMS = [
    # (clave, bloque, actividad, máximo)
    ('p1', 'Participación', 'Sesión 1 · Predicción: el DOM', 0.10),
    ('p2', 'Participación', 'Sesión 1 · Predicción: los eventos', 0.10),
    ('t1', 'Participación', 'Sesión 1 · Ticket de salida', 0.30),
    ('p3', 'Participación', 'Sesión 2 · Predicción: formularios', 0.10),
    ('p4', 'Participación', 'Sesión 2 · Predicción: datos', 0.10),
    ('t2', 'Participación', 'Sesión 2 · Ticket de salida', 0.30),
    ('r', 'Reto R5.1', 'Reto R5.1 · app.js de tu ERP (partes A, B, C y D y revisión final)', 9.00),
]
_MAX = {c: m for c, _, _, m in _ITEMS}
_NOMBRE_CALIF = 'calificacion_s5.json'
_CALIF = {}   # clave -> fracción lograda (0 a 1), mejor intento


def _total_posible():
    return round(sum(_MAX.values()), 2)


def _puntos(claves=None):
    return sum(_MAX[c] * _CALIF.get(c, 0) for c in (claves or _MAX))


def _guardar_calif():
    ruta = os.path.join(RUTA_PROYECTO, 'docs', _NOMBRE_CALIF)
    os.makedirs(os.path.dirname(ruta), exist_ok=True)
    with open(ruta, 'w', encoding='utf-8') as f:
        json.dump({'fracciones': _CALIF, 'actualizado': f"{datetime.now():%Y-%m-%d %H:%M}"}, f, indent=1)


def _cargar_calif():
    datos = leer_json(_NOMBRE_CALIF, {}) or {}
    for clave, fraccion in (datos.get('fracciones') or {}).items():
        if clave in _MAX:
            _CALIF[clave] = max(0.0, min(1.0, float(fraccion)))
    if _CALIF:
        print(f"📈 Recuperé tu avance anterior: {_puntos():.2f} / {_total_posible():g} puntos.")


def calif_registrar(clave, fraccion):
    """Guarda tu mejor intento de una actividad y muestra tu calificación acumulada."""
    if clave not in _MAX:
        return
    fraccion = max(0.0, min(1.0, float(fraccion)))
    _CALIF[clave] = max(_CALIF.get(clave, 0.0), round(fraccion, 4))
    _guardar_calif()
    actividad = next(a for c, _, a, _ in _ITEMS if c == clave)
    print(f"🧮 {actividad}: {_MAX[clave] * fraccion:.2f} de {_MAX[clave]:.2f} pts"
          f"  (tu mejor intento cuenta: {_MAX[clave] * _CALIF[clave]:.2f})")
    print(f"📈 CALIFICACIÓN ACUMULADA: {_puntos():.2f} / {_total_posible():g}")


# Las herramientas de la semana 1 ahora también registran puntos (si les pasas la clave)
_predecir_base = predecir
_autoverificar_base = autoverificar
_ticket_base = ticket_salida


def predecir(codigo, mi_prediccion, clave=None):
    _predecir_base(codigo, mi_prediccion)
    if clave and str(mi_prediccion).strip():
        calif_registrar(clave, 1.0)


def autoverificar(titulo, pruebas, meta=None, clave=None):
    porcentaje = _autoverificar_base(titulo, pruebas, meta)
    if clave:
        # Regla de puntos: se otorgan cuando el archivo existe y ya no queda ningún TODO
        bloqueo = []
        for descripcion, prueba, _ in pruebas:
            if descripcion.startswith(('No quedan', 'El archivo', 'Existen')):
                try:
                    ok = bool(prueba())
                except Exception:
                    ok = False
                if not ok:
                    bloqueo.append(descripcion)
        if bloqueo:
            print("🔒 Aún no sumas puntos: primero completa «" + bloqueo[0] + "» (guarda el archivo y quita todos los TODO).")
            calif_registrar(clave, 0)
        else:
            calif_registrar(clave, porcentaje / 100)
    return porcentaje


def ticket_salida(sesion, aprendi, me_bloqueo):
    if len(aprendi.strip()) < 15:
        print("✏️ Escribe con tus palabras qué aprendiste hoy (al menos una frase) y vuelve a ejecutar.")
        return
    _ticket_base(sesion, aprendi, me_bloqueo)
    calif_registrar('t' + str(sesion).split('.')[-1], 1.0)


def calificacion():
    """Tabla con tu calificación acumulada."""
    filas, bloque = '', None
    for clave, grupo, actividad, maximo in _ITEMS:
        if grupo != bloque:
            bloque = grupo
            pts_b = _puntos([c for c, g, _, _ in _ITEMS if g == grupo])
            max_b = sum(m for c, g, _, m in _ITEMS if g == grupo)
            filas += (f'<tr><td colspan="3" style="background:#edf2f7;font-weight:bold;padding:6px 10px">{grupo}'
                      f'</td><td style="background:#edf2f7;font-weight:bold;text-align:right;padding:6px 10px">'
                      f'{pts_b:.2f} / {max_b:.2f}</td></tr>')
        f = _CALIF.get(clave, 0)
        icono = '✅' if f >= 0.999 else ('🔶' if f > 0 else '⬜')
        barra = (f'<div style="background:#e2e8f0;border-radius:4px;height:10px;width:110px">'
                 f'<div style="background:{"#2f855a" if f >= 0.999 else "#b7791f"};height:10px;border-radius:4px;'
                 f'width:{int(110 * f)}px"></div></div>')
        filas += (f'<tr><td style="padding:4px 10px">{icono}</td><td style="padding:4px 10px">{actividad}</td>'
                  f'<td style="padding:4px 10px">{barra}</td>'
                  f'<td style="padding:4px 10px;text-align:right">{_MAX[clave] * f:.2f} / {_MAX[clave]:.2f}</td></tr>')
    total = _puntos()
    display(HTML(
        '<div style="font-family:system-ui,sans-serif;color:#1a202c;max-width:760px">'
        '<table style="border-collapse:collapse;width:100%;font-size:14px">'
        '<caption style="text-align:left;font-weight:bold;font-size:16px;padding:6px 0">📈 Mi calificación acumulada · Semana 5</caption>'
        f'<thead><tr><th scope="col"></th><th scope="col" style="text-align:left;padding:4px 10px">Actividad</th>'
        f'<th scope="col" style="text-align:left;padding:4px 10px">Avance</th>'
        f'<th scope="col" style="text-align:right;padding:4px 10px">Puntos</th></tr></thead>'
        f'<tbody>{filas}</tbody>'
        f'<tfoot><tr><th scope="row" colspan="3" style="text-align:right;padding:8px 10px;font-size:16px">TOTAL</th>'
        f'<td style="text-align:right;padding:8px 10px;font-size:18px;font-weight:bold;color:#1a365d">'
        f'{total:.2f} / {_total_posible():g}</td></tr></tfoot></table></div>'))
    return round(total, 2)


# ----------------------------------------------------------------------------------------------
# Boleta de calificación (se queda guardada en el cuaderno que entregas por Classroom)
# ----------------------------------------------------------------------------------------------
_SAL = 'cetis125-erp-s5-2026'
_ARCHIVOS_ENTREGA = ['app.js', 'estilos.css', 'index.html', 'inventario.html', 'formularios.html', 'ventas.html', 'clientes.html']


def _datos_boleta():
    return '|'.join(f"{c}:{_CALIF.get(c, 0):.4f}" for c in _MAX)


def _codigo_boleta(alumno, grupo, datos):
    base = f"{_SAL}|{alumno.strip().lower()}|{grupo}|{datos}"
    return hashlib.sha256(base.encode('utf-8')).hexdigest()[:10].upper()


def boleta():
    """Genera tu boleta final. Ejecútala al terminar y guarda el cuaderno con Ctrl+S."""
    if len(ALUMNO.strip()) < 5:
        print("⚠️ Escribe tu nombre completo en el campo ALUMNO de la CELDA 0, ejecútala de nuevo y regresa aquí.")
        return
    total = round(_puntos(), 2)
    datos = _datos_boleta()
    codigo = _codigo_boleta(ALUMNO, GRUPO, datos)
    ahora = f"{datetime.now():%d/%m/%Y %H:%M}"
    entregas = ''
    for nombre in _ARCHIVOS_ENTREGA:
        ruta = os.path.join(RUTA_PROYECTO, nombre)
        if os.path.exists(ruta):
            texto = open(ruta, encoding='utf-8').read()
            estado = 'con TODO pendientes' if 'TODO' in texto else 'completo'
            entregas += f'<li>📄 {nombre} · {len(texto)} caracteres · {estado}</li>'
        else:
            entregas += f'<li>⬜ {nombre} · no encontrado</li>'
    veredicto = ('🏆 Excelente' if total >= 9 else '👍 Aprobado' if total >= 6 else '🔧 Te falta avanzar: repite los autoverificadores')
    filas = ''
    for clave, grupo, actividad, maximo in _ITEMS:
        filas += (f'<tr><td style="padding:3px 8px">{grupo.split(" · ")[0]}</td><td style="padding:3px 8px">{actividad}</td>'
                  f'<td style="padding:3px 8px;text-align:right">{maximo * _CALIF.get(clave, 0):.2f} / {maximo:.2f}</td></tr>')
    html = (
        '<div style="font-family:system-ui,sans-serif;color:#1a202c;max-width:760px;border:3px solid #1a365d;'
        'border-radius:12px;padding:18px;background:#fff">'
        '<h2 style="margin:0 0 4px;color:#1a365d">📊 Boleta de calificación · Semana 5</h2>'
        '<p style="margin:0 0 10px">CETIS 125 · Submódulo 1: Construye aplicaciones web · JavaScript y DOM: tu ERP cobra vida</p>'
        f'<p style="margin:2px 0"><b>Alumno:</b> {html_lib.escape(ALUMNO.strip())} · <b>Grupo:</b> {GRUPO}</p>'
        f'<p style="margin:2px 0"><b>ERP:</b> {html_lib.escape(NOMBRE_ERP)} · <b>Fecha:</b> {ahora}</p>'
        f'<table style="border-collapse:collapse;width:100%;font-size:13px;margin:10px 0">'
        '<caption style="text-align:left;font-weight:bold">Detalle de puntos</caption>'
        '<thead><tr><th scope="col" style="text-align:left;padding:3px 8px">Bloque</th>'
        '<th scope="col" style="text-align:left;padding:3px 8px">Actividad</th>'
        '<th scope="col" style="text-align:right;padding:3px 8px">Puntos</th></tr></thead>'
        f'<tbody>{filas}</tbody></table>'
        f'<p style="font-size:24px;margin:8px 0"><b>Calificación formativa: {total:.2f} / 10</b> · {veredicto}</p>'
        f'<details open><summary><b>Archivos de tu proyecto</b></summary><ul style="margin:4px 0">{entregas}</ul></details>'
        f'<p style="font-size:12px;color:#4a5568;margin:10px 0 0">Código de verificación: <b>{codigo}</b><br>'
        f'<span style="word-break:break-all">Datos: {datos}</span></p></div>')
    display(HTML(html))
    with open(os.path.join(RUTA_PROYECTO, 'docs', 'boleta_s5.html'), 'w', encoding='utf-8') as f:
        f.write(html)
    print("💾 Copia guardada en tu proyecto: docs/boleta_s5.html")
    print("📌 AHORA: guarda el cuaderno con Ctrl+S (o Archivo → Guardar) y compártelo por Classroom.")
    return total


def verificar_boleta(alumno, grupo, datos, codigo):
    """(Docente) Comprueba que una boleta no fue alterada y recalcula la calificación."""
    esperado = _codigo_boleta(alumno, grupo, datos)
    total = 0.0
    for par in datos.split('|'):
        c, f = par.split(':')
        total += _MAX.get(c, 0) * float(f)
    if esperado == codigo.strip().upper():
        print(f"✅ Boleta auténtica. Calificación recalculada: {total:.2f} / {_total_posible():g}")
    else:
        print(f"❌ El código no coincide (esperado {esperado}). La boleta pudo ser modificada.")
    return round(total, 2)


# ----------------------------------------------------------------------------------------------
# Herramientas de análisis de HTML (semanas 3 a 5)
# ----------------------------------------------------------------------------------------------
def leer_pagina(nombre):
    ruta = os.path.join(RUTA_PROYECTO, nombre)
    if not os.path.exists(ruta):
        return ''
    with open(ruta, encoding='utf-8') as f:
        return f.read()


def contar(codigo_html, etiqueta):
    return [t for t, _ in _analizar(codigo_html).etiquetas].count(etiqueta)


def semanticas(codigo_html):
    etiquetas = [t for t, _ in _analizar(codigo_html).etiquetas]
    return sum(etiquetas.count(t) for t in REGIONES)


def texto_de(codigo_html, etiqueta):
    m = re.search(rf'<{etiqueta}\b[^>]*>(.*?)</{etiqueta}>', codigo_html, re.S | re.I)
    return ' '.join(re.sub(r'<[^>]+>', ' ', m.group(1)).split()) if m else ''


def enlaces(codigo_html):
    return [a.get('href', '') for t, a in _analizar(codigo_html).etiquetas if t == 'a']


def criterios_html(codigo_html):
    """Devuelve (aprobados, total) de revisar_html sin imprimir nada."""
    with contextlib.redirect_stdout(io.StringIO()) as buf:
        revisar_html(codigo_html)
    m = re.search(r'Resultado: (\d+)/(\d+)', buf.getvalue())
    return (int(m.group(1)), int(m.group(2))) if m else (0, 18)


def sin_divitis(codigo_html):
    return semanticas(codigo_html) >= 1 and contar(codigo_html, 'div') <= semanticas(codigo_html)


class _Tablas(HTMLParser):
    def __init__(self):
        super().__init__()
        self.tablas, self._t, self._fila, self._sec, self._cap = [], None, None, None, False

    def handle_starttag(self, tag, attrs):
        a = {k: (v or '') for k, v in attrs}
        if tag == 'table':
            self._t = {'caption': '', 'thead': False, 'tbody': False, 'tfoot': False, 'th': [], 'filas': [],
                       'maquillaje': [k for k in a if k in ('border', 'bgcolor', 'align', 'cellpadding', 'cellspacing', 'width')]}
            self.tablas.append(self._t)
        elif self._t is None:
            return
        elif tag == 'caption':
            self._cap = True
        elif tag in ('thead', 'tbody', 'tfoot'):
            self._t[tag] = True
            self._sec = tag
        elif tag == 'tr':
            self._fila = {'sec': self._sec, 'celdas': 0, 'th': 0, 'td': 0}
            self._t['filas'].append(self._fila)
        elif tag in ('th', 'td') and self._fila is not None:
            self._fila['celdas'] += 1
            self._fila[tag] += 1
            if tag == 'th':
                self._t['th'].append({'scope': a.get('scope', ''), 'sec': self._sec})

    def handle_endtag(self, tag):
        if tag == 'caption':
            self._cap = False
        elif tag in ('thead', 'tbody', 'tfoot'):
            self._sec = None
        elif tag == 'tr':
            self._fila = None
        elif tag == 'table':
            self._t = None

    def handle_data(self, data):
        if self._cap and self._t is not None:
            self._t['caption'] += data


def tablas(codigo_html):
    p = _Tablas()
    p.feed(codigo_html)
    return p.tablas


def tabla_envuelta(codigo_html):
    """¿La tabla vive dentro de un contenedor con overflow-x: auto?"""
    return bool(re.search(r'overflow-x\s*:\s*(auto|scroll)', codigo_html)) and \
        bool(re.search(r'<(div|section)[^>]*class="[^"]*(scroll|tabla)[^"]*"[^>]*>\s*<table', codigo_html, re.S))


def analizar_tablas(codigo_html, nombre='la página'):
    print(f"📊 Radiografía de las tablas de {nombre}")
    print("-" * 75)
    encontradas = tablas(codigo_html)
    if not encontradas:
        print("⚠️ No encontré ninguna <table>.")
    for i, t in enumerate(encontradas, 1):
        cols = max([f['celdas'] for f in t['filas']] or [0])
        cuerpo = [f for f in t['filas'] if f['sec'] == 'tbody']
        print(f"Tabla {i}: {len(t['filas'])} filas · {cols} columnas")
        print(f"   {'✅' if t['caption'].strip() else '❌'} <caption>: {' '.join(t['caption'].split()) or '(falta)'}")
        print(f"   {'✅' if t['thead'] else '❌'} <thead>    {'✅' if t['tbody'] else '❌'} <tbody>    {'✅' if t['tfoot'] else '➖'} <tfoot>")
        con_scope = sum(1 for h in t['th'] if h['scope'] in ('col', 'row'))
        print(f"   {'✅' if t['th'] and con_scope == len(t['th']) else '❌'} <th> con scope: {con_scope} de {len(t['th'])}")
        if t['maquillaje']:
            print(f"   ⚠️ Atributos viejos de maquillaje: {', '.join(t['maquillaje'])} (el aspecto va en CSS)")
        if '<html' in codigo_html.lower():
            print(f"   📦 Contenedor con scroll horizontal: {'✅' if tabla_envuelta(codigo_html) else '❌'}")
    print("-" * 75)


class _Formularios(HTMLParser):
    def __init__(self):
        super().__init__()
        self.formularios, self._f, self._lab, self._btn = [], None, 0, None

    def handle_starttag(self, tag, attrs):
        a = {k: (v or '') for k, v in attrs}
        if tag == 'form':
            self._f = {'fieldset': 0, 'legend': 0, 'controles': [], 'labels_for': set(), 'botones': []}
            self.formularios.append(self._f)
            return
        if self._f is None:
            return
        if tag in ('fieldset', 'legend'):
            self._f[tag] += 1
        elif tag == 'label':
            self._lab += 1
            if a.get('for'):
                self._f['labels_for'].add(a['for'])
        elif tag in ('input', 'select', 'textarea'):
            self._f['controles'].append({'tag': tag, 'a': a, 'envuelto': self._lab > 0})
        elif tag == 'button':
            self._btn = {'a': a, 'texto': ''}

    def handle_endtag(self, tag):
        if tag == 'label':
            self._lab = max(0, self._lab - 1)
        elif tag == 'button' and self._btn is not None and self._f is not None:
            self._f['botones'].append(self._btn)
            self._btn = None
        elif tag == 'form':
            self._f = None

    def handle_data(self, data):
        if self._btn is not None:
            self._btn['texto'] += data


def formularios(codigo_html):
    p = _Formularios()
    p.feed(codigo_html)
    for f in p.formularios:
        f['campos'] = [c for c in f['controles'] if c['a'].get('type') not in ('hidden', 'submit', 'button', 'reset', 'image')]
        for c in f['campos']:
            c['etiqueta'] = bool(c['envuelto'] or c['a'].get('id') in f['labels_for'] or c['a'].get('aria-label'))
    return p.formularios


def analizar_formularios(codigo_html, nombre='la página'):
    print(f"📝 Radiografía de los formularios de {nombre}")
    print("-" * 75)
    encontrados = formularios(codigo_html)
    if not encontrados:
        print("⚠️ No encontré ningún <form>.")
    for i, f in enumerate(encontrados, 1):
        print(f"Formulario {i}: {len(f['campos'])} campos · fieldset: {f['fieldset']} · legend: {f['legend']} · botones: {len(f['botones'])}")
        for c in f['campos']:
            a = c['a']
            tipo = a.get('type', 'text') if c['tag'] == 'input' else c['tag']
            reglas = ', '.join(r for r in ('required', 'min', 'max', 'minlength', 'maxlength', 'pattern') if r in a) or '—'
            print(f"   {'✅' if c['etiqueta'] else '❌'} {a.get('name') or a.get('id') or '(sin nombre)':<14} tipo: {tipo:<9} label: {'sí' if c['etiqueta'] else 'NO'}   reglas: {reglas}")
    print("-" * 75)


# ----------------------------------------------------------------------------------------------
# Menú común del ERP (consistencia: Nielsen #4)
# ----------------------------------------------------------------------------------------------
PAGINAS_ERP = [('index.html', 'Tablero'), ('inventario.html', 'Inventario'), ('ventas.html', 'Ventas'), ('clientes.html', 'Clientes')]


def menu_erp(activa):
    items = ''
    for archivo, texto in PAGINAS_ERP:
        actual = ' aria-current="page"' if archivo == activa else ''
        items += f'<li><a href="{archivo}"{actual}>{texto}</a></li>\n    '
    return f'<nav aria-label="Principal">\n  <ul>\n    {items.rstrip()}\n  </ul>\n</nav>'


def poner_menu(archivo):
    """Reemplaza el <nav> de un archivo por el menú común del ERP."""
    codigo = leer_pagina(archivo)
    if not re.search(r'<nav\b.*?</nav>', codigo, re.S):
        print(f"⚠️ {archivo} no tiene <nav>; agrégalo con el menú: {menu_erp(archivo)}")
        return
    nuevo = re.sub(r'<nav\b.*?</nav>', lambda m: menu_erp(archivo), codigo, count=1, flags=re.S)
    guardar(archivo, nuevo)



# ----------------------------------------------------------------------------------------------
# Utilidades de estilos y de páginas (semanas 3 y 4)
# ----------------------------------------------------------------------------------------------
_ESTILOS_BASE = r'''
:root {
  --color-fondo: #f7fafc; --color-texto: #1a202c; --color-primario: #1a365d; --color-sobre-primario: #ffffff;
  --color-borde: #718096; --espacio: 1rem; --fuente: system-ui, sans-serif;
}
*, *::before, *::after { box-sizing: border-box; }
body { margin: 0; font-family: var(--fuente); font-size: 1rem; line-height: 1.6; color: var(--color-texto); background: var(--color-fondo);
  display: flex; flex-direction: column; min-height: 100vh; }
header { background: var(--color-primario); color: var(--color-sobre-primario); padding: var(--espacio); }
nav ul { list-style: none; margin: 0; padding: var(--espacio); display: flex; flex-wrap: wrap; gap: .5rem; }
nav a { display: inline-flex; align-items: center; min-height: 2.75rem; padding: 0 1rem; }
main { flex: 1; padding: var(--espacio); display: grid; gap: var(--espacio); grid-template-columns: minmax(0, 1fr); }
@media (min-width: 48rem) { main { grid-template-columns: repeat(2, minmax(0, 1fr)); } }
footer { background: var(--color-primario); color: var(--color-sobre-primario); padding: var(--espacio); text-align: center; }
:focus-visible { outline: 3px solid var(--color-primario); outline-offset: 2px; }
th, td { border: 1px solid var(--color-borde); padding: .5rem; }
.tabla-scroll { overflow-x: auto; }
img, video { max-width: 100%; height: auto; }
'''


def bloque_css(nombre, css, archivo='estilos.css'):
    """Agrega (o reemplaza) un bloque con nombre dentro de estilos.css: puedes volver a ejecutar la celda sin duplicar."""
    inicio, fin = f"/* ===== BLOQUE {nombre} ===== */", f"/* ===== FIN {nombre} ===== */"
    actual = leer_pagina(archivo)
    actual = re.sub(re.escape(inicio) + r'.*?' + re.escape(fin) + r'\n?', '', actual, flags=re.S).rstrip()
    guardar(archivo, actual + f"\n\n{inicio}\n{css.strip()}\n{fin}\n")


def poner_estilos(archivo, css='estilos.css'):
    """Agrega <link rel="stylesheet" href="estilos.css"> al <head> de una página (si no lo tiene)."""
    codigo = leer_pagina(archivo)
    if not codigo:
        print(f"⚠️ {archivo} no existe todavía.")
        return
    if enlaza_css(codigo, css):
        print(f"✅ {archivo} ya enlaza {css}")
        return
    if '</head>' not in codigo:
        print(f"⚠️ {archivo} no tiene </head>.")
        return
    guardar(archivo, codigo.replace('</head>', f'  <link rel="stylesheet" href="{css}">\n</head>', 1))


def poner_estilos_todas():
    for archivo in _ARCHIVOS_HTML:
        poner_estilos(archivo)


def enlaza_css(codigo_html, css='estilos.css'):
    return any(t == 'link' and a.get('rel', '').lower() == 'stylesheet' and a.get('href', '').split('?')[0] == css
               for t, a in _analizar(codigo_html).etiquetas)


def pagina_con_estilos(archivo_o_html, css='estilos.css'):
    """Devuelve la página con su hoja de estilos incrustada (para verla dentro de Colab)."""
    codigo = leer_pagina(archivo_o_html) if archivo_o_html.endswith('.html') and '<' not in archivo_o_html else archivo_o_html
    hoja = leer_pagina(css)
    codigo = re.sub(r'<link[^>]*rel="stylesheet"[^>]*>', '', codigo)
    return codigo.replace('</head>', f'<style>\n{hoja}\n</style>\n</head>', 1)


def ver(archivo, anchos=(360, 768, 1280), alto=520):
    """Muestra una página de tu proyecto con estilos en varios anchos de pantalla."""
    if not leer_pagina(archivo):
        print(f"⚠️ {archivo} no existe. Ejecuta rescatar_semana4() si te faltan páginas.")
        return
    comparar_pantallas(pagina_con_estilos(archivo), anchos=anchos, alto=alto)


# --- Páginas de la semana 3 (por si falta alguna) --------------------------------------------------
_ARCHIVOS_HTML = ['index.html', 'inventario.html', 'ventas.html', 'clientes.html', 'formularios.html']


def _pagina_base(archivo):
    titulos = {'index.html': 'Tablero', 'inventario.html': 'Inventario', 'ventas.html': 'Ventas',
               'clientes.html': 'Clientes', 'formularios.html': 'Captura de datos'}
    cabecera = (f'<!DOCTYPE html>\n<html lang="es">\n<head>\n  <meta charset="UTF-8">\n'
                f'  <meta name="viewport" content="width=device-width, initial-scale=1.0">\n'
                f'  <title>{titulos[archivo]} · {NOMBRE_ERP}</title>\n</head>\n<body>\n'
                f'  <header>\n    <h1>{NOMBRE_ERP if archivo == "index.html" else titulos[archivo]}</h1>\n  </header>\n'
                f'  {menu_erp(archivo)}\n  <main>\n')
    pie = f'  </main>\n  <footer>\n    <p>{NOMBRE_ERP} · Equipo: {", ".join(INTEGRANTES)} · CETIS 125</p>\n  </footer>\n</body>\n</html>'

    def tabla(titulo, columnas, filas):
        enc = ''.join(f'<th scope="col">{c}</th>' for c in columnas)
        cuerpo = '\n'.join('          <tr>' + ''.join(f'<td>{x}</td>' for x in f) + '</tr>' for f in filas)
        return (f'    <section>\n      <h2>{titulo}</h2>\n      <div class="tabla-scroll">\n        <table>\n'
                f'          <caption>{titulo}</caption>\n          <thead><tr>{enc}</tr></thead>\n'
                f'          <tbody>\n{cuerpo}\n          </tbody>\n        </table>\n      </div>\n    </section>\n')

    if archivo == 'index.html':
        cuerpo = ''.join(f'    <section id="{i}">\n      <h2>{t}</h2>\n      <article><h3>{a}</h3><p>{p}</p></article>\n'
                         f'      <article><h3>{a2}</h3><p>{p2}</p></article>\n    </section>\n'
                         for i, t, a, p, a2, p2 in (('ventas', 'Ventas de hoy', 'Total vendido', '$1,240', 'Tickets', '18'),
                                                    ('stock', 'Stock bajo', 'Por reponer', '5 productos', 'Críticos', '2'),
                                                    ('modulo', 'Pendientes', 'Por cobrar', '3 clientes', 'Pedidos', '4')))
    elif archivo == 'inventario.html':
        cuerpo = tabla('Productos en existencia', ['Código', 'Producto', 'Existencia', 'Precio', 'Proveedor'],
                       [['P001', 'Café molido 250 g', '12', '$85.00', 'Cafés del Sur'], ['P002', 'Azúcar 1 kg', '30', '$32.50', 'Abarrotes Rosa'],
                        ['P003', 'Leche 1 L', '18', '$27.00', 'Lácteos Norte'], ['P004', 'Pan de caja', '9', '$45.00', 'Panadería Luz']])
    elif archivo == 'ventas.html':
        cuerpo = tabla('Ventas del día', ['Folio', 'Cliente', 'Producto', 'Cantidad', 'Total'],
                       [['V001', 'Ana', 'Café molido', '2', '$170.00'], ['V002', 'Luis', 'Azúcar 1 kg', '1', '$32.50'],
                        ['V003', 'Marta', 'Leche 1 L', '3', '$81.00']]) + \
            '    <p><a href="formularios.html">➕ Registrar una nueva venta</a></p>\n'
    elif archivo == 'clientes.html':
        cuerpo = tabla('Clientes registrados', ['Clave', 'Nombre', 'Teléfono', 'Adeudo'],
                       [['C01', 'Ana Pérez', '555 123 4567', '$0.00'], ['C02', 'Luis Mora', '555 987 6543', '$120.00'],
                        ['C03', 'Marta Ríos', '555 456 7890', '$0.00']])
    else:
        cuerpo = ('    <section>\n      <h2>Nuevo producto</h2>\n      <form>\n        <fieldset>\n          <legend>Datos del producto</legend>\n'
                  '          <label for="prod-codigo">Código</label>\n          <input id="prod-codigo" name="codigo" required pattern="[A-Z][0-9]{3}">\n'
                  '          <label for="prod-nombre">Nombre</label>\n          <input id="prod-nombre" name="nombre" required>\n'
                  '          <label for="prod-existencia">Existencia</label>\n          <input id="prod-existencia" name="existencia" type="number" min="0" required>\n'
                  '          <label for="prod-precio">Precio</label>\n          <input id="prod-precio" name="precio" type="number" min="0" step="0.01" required>\n'
                  '          <label for="prod-categoria">Categoría</label>\n          <select id="prod-categoria" name="categoria"><option>Abarrotes</option><option>Bebidas</option></select>\n'
                  '        </fieldset>\n        <button type="submit">Guardar producto</button>\n      </form>\n    </section>\n'
                  '    <section>\n      <h2>Nueva venta</h2>\n      <form>\n        <fieldset>\n          <legend>Datos de la venta</legend>\n'
                  '          <label for="venta-producto">Producto</label>\n          <select id="venta-producto" name="producto"><option>Café molido</option><option>Azúcar 1 kg</option></select>\n'
                  '          <label for="venta-cantidad">Cantidad</label>\n          <input id="venta-cantidad" name="cantidad" type="number" min="1" required>\n'
                  '          <label for="venta-pago">Forma de pago</label>\n          <select id="venta-pago" name="pago"><option>Efectivo</option><option>Tarjeta</option></select>\n'
                  '          <label for="venta-fecha">Fecha</label>\n          <input id="venta-fecha" name="fecha" type="date" required>\n'
                  '        </fieldset>\n        <button type="submit">Cobrar</button>\n      </form>\n    </section>\n')
    return cabecera + cuerpo + pie






def rescatar_semana4():
    """Si te falta alguna página (o tu estilos.css), crea una versión base para poder seguir con JavaScript."""
    creadas = []
    for archivo in _ARCHIVOS_HTML:
        if not leer_pagina(archivo).strip():
            guardar(archivo, _pagina_base(archivo))
            creadas.append(archivo)
    if not leer_pagina('estilos.css').strip():
        guardar('estilos.css', _ESTILOS_BASE.strip() + '\n')
        creadas.append('estilos.css')
    poner_estilos_todas()
    if creadas:
        print(f"🛟 Creé estos archivos base: {', '.join(creadas)}. Cámbialos luego por los de tu negocio.")
    else:
        print("✅ Ya tienes tus 5 páginas y tu estilos.css.")


rescatar_semana3 = rescatar_semana4


# ----------------------------------------------------------------------------------------------
# Herramientas de JavaScript (semana 5): Node + jsdom para ejecutar y probar tus páginas
# ----------------------------------------------------------------------------------------------
import subprocess
import shutil
import tempfile

RUTA_JS = os.environ.get('ERP_JS_DIR') or ('/content/_erp_js' if os.path.isdir('/content') else os.path.join(tempfile.gettempdir(), '_erp_js'))

_RUNNER_JS = r"""
const { JSDOM, VirtualConsole } = require('jsdom');
const vm = require('vm');
const util = require('util');
const fs = require('fs');
const entrada = JSON.parse(fs.readFileSync(0, 'utf8'));

const AYUDAS = `
window.$ = (s, r) => (r || document).querySelector(s);
window.$$ = (s, r) => Array.from((r || document).querySelectorAll(s));
window.esperar = (ms) => new Promise((ok) => setTimeout(ok, ms || 20));
window.escribir = (el, v) => { el.value = v; el.dispatchEvent(new Event('input', { bubbles: true })); el.dispatchEvent(new Event('change', { bubbles: true })); };
window.clic = (el) => el.click();
window.enviar = (f) => { window.__evento = null; f.addEventListener('submit', (e) => { window.__evento = e; }, { once: true }); f.requestSubmit(); return window.__evento; };
window.texto = (el) => (el ? el.textContent.replace(/\\s+/g, ' ').trim() : '');
window.numero = (t) => Number(String(t).replace(/[^0-9.\\-]/g, ''));
`;

async function probar(prueba) {
  const errores = [];
  const vc = new VirtualConsole();
  vc.on('jsdomError', (e) => errores.push(String((e && e.detail && e.detail.message) || (e && e.message) || e)));
  vc.on('error', (...a) => errores.push('console.error: ' + a.join(' ')));
  const dom = new JSDOM(prueba.html, {
    runScripts: 'dangerously', url: 'http://localhost/', pretendToBeVisual: true, virtualConsole: vc,
    beforeParse(w) { for (const k of Object.keys(prueba.almacen || {})) w.localStorage.setItem(k, prueba.almacen[k]); }
  });
  const w = dom.window;
  if (w.document.readyState !== 'complete') await new Promise((ok) => w.addEventListener('load', ok));
  await new Promise((ok) => setTimeout(ok, 15));
  w.__errores = errores;
  w.eval(AYUDAS);
  let ok = false, error = '';
  try { ok = !!(await w.eval('(async () => {' + prueba.js + '\n})()')); }
  catch (e) { error = String((e && e.message) || e); }
  if (ok && prueba.js2) {
    // Segunda fase: «recargar» la página conservando lo guardado en localStorage
    const guardado = {};
    for (let i = 0; i < w.localStorage.length; i++) { const k = w.localStorage.key(i); guardado[k] = w.localStorage.getItem(k); }
    w.close();
    const errores2 = [];
    const vc2 = new VirtualConsole();
    vc2.on('jsdomError', (e) => errores2.push(String((e && e.detail && e.detail.message) || (e && e.message) || e)));
    vc2.on('error', (...a) => errores2.push('console.error: ' + a.join(' ')));
    const dom2 = new JSDOM(prueba.html, {
      runScripts: 'dangerously', url: 'http://localhost/', pretendToBeVisual: true, virtualConsole: vc2,
      beforeParse(w2) { for (const k of Object.keys(guardado)) w2.localStorage.setItem(k, guardado[k]); }
    });
    const w2 = dom2.window;
    if (w2.document.readyState !== 'complete') await new Promise((ok2) => w2.addEventListener('load', ok2));
    await new Promise((ok2) => setTimeout(ok2, 15));
    w2.__errores = errores2;
    w2.eval(AYUDAS);
    try { ok = !!(await w2.eval('(async () => {' + prueba.js2 + '\n})()')); }
    catch (e) { ok = false; error = String((e && e.message) || e); }
    w2.close();
    return { id: prueba.id, ok, error, errores: errores.concat(errores2) };
  }
  w.close();
  return { id: prueba.id, ok, error, errores };
}

(async () => {
  const salida = [];
  if (entrada.modo === 'consola') {
    const lineas = [];
    const consola = { log: (...a) => lineas.push(util.format(...a)), error: (...a) => lineas.push(util.format(...a)) };
    let error = '';
    try { vm.runInNewContext(entrada.codigo, { console: consola, Math, Number, JSON, Date, Intl, String, Array, Object, parseInt, parseFloat, isNaN }, { timeout: 2000 }); }
    catch (e) { error = String((e && e.message) || e); }
    process.stdout.write(JSON.stringify({ lineas, error }));
    return;
  }
  for (const p of entrada.pruebas) salida.push(await probar(p));
  process.stdout.write(JSON.stringify(salida));
})();
"""

_JS_LISTO = None


def _asegurar_node():
    """Prepara Node + jsdom (la primera vez tarda unos segundos)."""
    global _JS_LISTO
    if _JS_LISTO is not None:
        return _JS_LISTO
    if shutil.which('node') is None or shutil.which('npm') is None:
        print("⚠️ No encontré Node.js en este entorno; no puedo ejecutar las pruebas de JavaScript.")
        _JS_LISTO = False
        return False
    os.makedirs(RUTA_JS, exist_ok=True)
    if not os.path.isdir(os.path.join(RUTA_JS, 'node_modules', 'jsdom')):
        print("🔧 Preparando el probador de JavaScript (jsdom). Solo tarda la primera vez...")
        try:
            subprocess.run(['npm', 'install', 'jsdom', '--silent', '--no-audit', '--no-fund'],
                           cwd=RUTA_JS, check=True, timeout=300, capture_output=True)
        except Exception as e:
            print(f"⚠️ No pude instalar jsdom ({type(e).__name__}). Revisa tu conexión y ejecuta la CELDA 0 otra vez.")
            _JS_LISTO = False
            return False
    with open(os.path.join(RUTA_JS, 'runner.js'), 'w', encoding='utf-8') as f:
        f.write(_RUNNER_JS)
    _JS_LISTO = True
    return True


def _node(entrada, tiempo=120):
    pr = subprocess.run(['node', 'runner.js'], cwd=RUTA_JS, input=json.dumps(entrada), text=True,
                        capture_output=True, timeout=tiempo)
    if pr.returncode != 0:
        raise RuntimeError(pr.stderr.strip()[-400:] or 'Node falló')
    return json.loads(pr.stdout)


def correr_js(codigo):
    """Ejecuta un fragmento de JavaScript y devuelve lo que imprimió con console.log."""
    if not _asegurar_node():
        return None
    r = _node({'modo': 'consola', 'codigo': codigo}, 30)
    return '\n'.join(r['lineas']) + (('\n' + r['error']) if r['error'] else '')


def predecir_js(codigo, mi_prediccion, clave=None):
    """Como predecir(), pero el código es JavaScript."""
    if not str(mi_prediccion).strip():
        print("✏️ Primero escribe tu predicción en MI_PREDICCION y vuelve a ejecutar.")
        return
    real = correr_js(codigo)
    if real is None:
        print("🔌 No pude ejecutar JavaScript aquí; tu predicción quedó registrada.")
        real = str(mi_prediccion)
    real = real.strip()
    if ' '.join(str(mi_prediccion).split()) == ' '.join(real.split()):
        print(f"✅ ¡Acertaste! El código imprime: {real}")
    else:
        print(f"🤔 Tu predicción:  {mi_prediccion}")
        print(f"   Resultado real: {real}")
        print("   💬 Coméntalo con tu pareja: ¿en qué línea cambió lo que esperabas?")
    if clave:
        calif_registrar(clave, 1.0)


# --- Armar páginas con su JavaScript ----------------------------------------------------------
_RE_SCRIPT = re.compile(r'<script\b([^>]*?)\bsrc\s*=\s*"([^"]+)"([^>]*)>\s*</script>', re.I | re.S)


def _seguro(codigo_js):
    return codigo_js.replace('</script', '<\\/script')


def inlinar_js(codigo_html, js_extra=None, carpeta=None):
    """Sustituye <script src="x.js"> por el contenido de tu archivo (para probarlo o verlo en Colab).
    Los scripts con `defer` se ejecutan al final, igual que en un navegador."""
    carpeta = carpeta or RUTA_PROYECTO
    diferidos = []

    def cambiar(m):
        atributos, src = m.group(1) + ' ' + m.group(3), m.group(2)
        ruta = os.path.join(carpeta, src.split('?')[0])
        if src.startswith(('http:', 'https:', '//')) or not os.path.exists(ruta):
            return m.group(0)
        with open(ruta, encoding='utf-8') as f:
            contenido = f.read()
        if re.search(r'\bdefer\b|\btype\s*=\s*"module"', atributos, re.I):
            diferidos.append(contenido)
            return ''
        return f'<script>{_seguro(contenido)}</script>'

    codigo = _RE_SCRIPT.sub(cambiar, codigo_html)
    if js_extra:
        diferidos.append(js_extra)
    if diferidos:
        bloque = ''.join(f'<script>{_seguro(c)}</script>\n' for c in diferidos)
        codigo = codigo.replace('</body>', bloque + '</body>', 1) if '</body>' in codigo else codigo + bloque
    return codigo


def pagina_viva(archivo_o_html, js=None, css='estilos.css'):
    """La página con estilos y JavaScript incrustados, lista para mostrarla funcionando dentro de Colab."""
    if archivo_o_html.endswith('.html') and '<' not in archivo_o_html:
        codigo = pagina_con_estilos(archivo_o_html, css)
    else:
        codigo = archivo_o_html
    return inlinar_js(codigo, js)


def ver_pagina(archivo_o_html, js=None, alto=520):
    """Muestra tu página FUNCIONANDO (puedes escribir, dar clic y enviar formularios dentro del marco)."""
    if archivo_o_html.endswith('.html') and '<' not in archivo_o_html and not leer_pagina(archivo_o_html):
        print(f"⚠️ {archivo_o_html} no existe. Ejecuta rescatar_semana4() si te faltan páginas.")
        return
    mostrar(pagina_viva(archivo_o_html, js), alto)


def probar_js(fuente, pruebas, js=None, almacen=None):
    """Ejecuta pruebas en una página con jsdom.
    fuente: nombre de archivo de tu proyecto o código HTML · js: código extra (para los ejercicios guiados)
    pruebas: lista de (descripción, código_js_de_la_prueba, pista[, opciones]); opciones = {'paginas': [...], 'almacen': {...}, 'recargar': código JS que se ejecuta en la página recargada}
    Devuelve una lista de (ok, detalle)."""
    if not _asegurar_node():
        return None
    trabajos, destino = [], []
    for i, p in enumerate(pruebas):
        opciones = p[3] if len(p) > 3 else {}
        paginas = opciones.get('paginas') or [fuente]
        for pagina in paginas:
            codigo = leer_pagina(pagina) if (pagina.endswith('.html') and '<' not in pagina) else pagina
            if not codigo:
                destino.append((i, pagina, False, f'no existe {pagina}'))
                continue
            trabajos.append({'id': f'{i}|{pagina}', 'js': p[1], 'js2': opciones.get('recargar', ''), 'html': inlinar_js(codigo, js),
                             'almacen': opciones.get('almacen', almacen) or {}})
    salidas = _node({'pruebas': trabajos}) if trabajos else []
    por_prueba = {i: [] for i in range(len(pruebas))}
    for i, pagina, ok, detalle in destino:
        por_prueba[i].append((pagina, ok, detalle))
    for s in salidas:
        i, pagina = s['id'].split('|', 1)
        detalle = s['error'] or ('; '.join(s['errores'][:1]) if not s['ok'] and s['errores'] else '')
        por_prueba[int(i)].append((pagina, s['ok'], detalle))
    resultado = []
    for i in range(len(pruebas)):
        fallas = [(pg, d) for pg, ok, d in por_prueba[i] if not ok]
        if fallas:
            quien = ', '.join(pg for pg, _ in fallas if pg.endswith('.html'))
            detalle = fallas[0][1]
            resultado.append((False, ((f'falla en {quien}' if quien else '') + (f' · {detalle}' if detalle else '')).strip(' ·')))
        else:
            resultado.append((True, ''))
    return resultado


def autoverificar_js(titulo, fuente, pruebas, js=None, almacen=None, meta=None, clave=None):
    """Autoverificador de JavaScript. Cada prueba es (descripción, X, pista[, opciones]):
    si X es un texto, es código JS que se ejecuta DENTRO de tu página (devuelve true/false);
    si X es una función de Python, se evalúa directamente."""
    js_pruebas = [p for p in pruebas if isinstance(p[1], str)]
    resultados = probar_js(fuente, js_pruebas, js=js, almacen=almacen) if js_pruebas else []
    if resultados is None:
        print("🔌 No pude preparar el probador de JavaScript. Ejecuta la CELDA 0 otra vez y revisa tu conexión.")
        return 0

    def fijo(ok, detalle):
        def evaluar():
            if not ok:
                raise AssertionError(detalle or 'no se cumplió')
            return True
        return evaluar

    it = iter(resultados)
    compuestas = []
    for p in pruebas:
        d, x, h = p[0], p[1], p[2]
        if isinstance(x, str):
            ok, detalle = next(it)
            compuestas.append((d, fijo(ok, detalle), h))
        else:
            compuestas.append((d, x, h))
    return autoverificar(titulo, compuestas, meta=meta, clave=clave)


def sin_todo(codigo):
    return 'TODO' not in codigo


# --- app.js por bloques (igual que estilos.css) ------------------------------------------------
def bloque_js(nombre, codigo, archivo='app.js'):
    """Agrega (o reemplaza) un bloque con nombre dentro de app.js: puedes volver a ejecutar la celda sin duplicar."""
    inicio, fin = f"/* ===== BLOQUE {nombre} ===== */", f"/* ===== FIN {nombre} ===== */"
    actual = leer_pagina(archivo)
    actual = re.sub(re.escape(inicio) + r'.*?' + re.escape(fin) + r'\n?', '', actual, flags=re.S).rstrip()
    guardar(archivo, (actual + '\n\n' if actual else '') + f"{inicio}\n{codigo.strip()}\n{fin}\n")


def enlaza_js(codigo_html, js='app.js'):
    return any(t == 'script' and a.get('src', '').split('?')[0] == js for t, a in _analizar(codigo_html).etiquetas)


def poner_script(archivo, js='app.js'):
    """Agrega <script src="app.js" defer> al <head> de una página (si no lo tiene)."""
    codigo = leer_pagina(archivo)
    if not codigo:
        print(f"⚠️ {archivo} no existe todavía.")
        return
    if enlaza_js(codigo, js):
        print(f"✅ {archivo} ya enlaza {js}")
        return
    if '</head>' not in codigo:
        print(f"⚠️ {archivo} no tiene </head>.")
        return
    guardar(archivo, codigo.replace('</head>', f'  <script src="{js}" defer></script>\n</head>', 1))


def poner_script_todas():
    for archivo in _ARCHIVOS_HTML:
        poner_script(archivo)


def unir_pruebas(*listas):
    """Junta varias listas de pruebas en una sola (sin repetir descripciones)."""
    vistos, salida = set(), []
    for lista in listas:
        for prueba in lista:
            if prueba[0] not in vistos:
                vistos.add(prueba[0])
                salida.append(prueba)
    return salida


_cargar_calif()


print("=" * 75)
print(f"📁 Carpeta del proyecto:\n{RUTA_PROYECTO}")
print(f"🏪 ERP: {NOMBRE_ERP}" + ("" if ALUMNO.strip() else "   ⚠️ Escribe tu nombre en ALUMNO y ejecuta esta celda otra vez"))
print("🧰 Herramientas de siempre: predecir, autoverificar, guardar, leer, mostrar, comparar_pantallas,")
print("   revisar_html, menu_erp, ticket_salida, registrar_ia, arbol, calificacion y boleta")
print("   NUEVAS (semana 5): predecir_js, autoverificar_js, ver_pagina (tu página funcionando),")
print("   bloque_js (arma app.js por partes), poner_script_todas, unir_pruebas, rescatar_semana4")
_asegurar_node()
print("=" * 75)
````

### 2.1 · Puente Python → JavaScript

### 2.2 · El DOM: tu página convertida en un árbol de objetos
Cuando abres una página, el navegador lee tu HTML y construye el **DOM** (*Document Object Model*): un **árbol** donde **cada etiqueta es un objeto** que JavaScript puede **buscar, leer y modificar** *(MDN, Introducción al DOM)*.

```
document
└── html
    ├── head ── title
    └── body
        ├── header ── h1
        ├── nav ── ul ── li ── a
        ├── main ── section ── table ── tbody ── tr ── td
        └── footer ── p
```

Cambiar el DOM **no cambia tu archivo `.html`**: cambia lo que ve la persona *en este momento*. Si recarga la página, el navegador vuelve a construir el árbol desde el HTML (por eso en la sesión 2 aprenderás a **guardar** datos).

**¿Cómo se conecta tu JavaScript con la página?** Con una etiqueta `<script>` en el `<head>`:
```html
<script src="app.js" defer></script>
```
`defer` le dice al navegador: *“descarga el script ya, pero ejecútalo cuando el HTML esté completo”*. Sin `defer`, el script corre cuando todavía no existen los elementos y `document.querySelector(...)` devuelve `null`. Por eso la celda **Paso 0** puso esa etiqueta en tus páginas.

**La consola es tu mejor amiga.** Los mensajes de `console.log(...)` y **todos los errores** aparecen en la consola del navegador (tecla **F12** → pestaña *Consola*). Un error sale en rojo y dice **qué falló y en qué línea**. En Colab, `ver_pagina(...)` te muestra la página funcionando y el autoverificador te dice qué prueba falló; si algo no hace nada, lo primero es buscar el mensaje rojo.

| Si la consola dice… | Casi siempre significa… |
|---|---|
| `Cannot read properties of null (reading 'addEventListener')` | Tu `querySelector` no encontró nada: revisa el selector (`#`, `.`) o que el script lleve `defer` |
| `x is not defined` | Escribiste mal el nombre de una variable o función |
| `Unexpected token` / `missing )` | Falta o sobra un paréntesis, llave o comilla |
| `Assignment to constant variable` | Cambiaste el valor de algo declarado con `const` (usa `let`) |

### 2.3 · Buscar, leer y cambiar elementos
El trabajo con el DOM siempre tiene **tres pasos**: **① buscar** el elemento → **② cambiarlo** → **③ comprobar** (con `console.log` o mirando la página).

| Quiero… | Código | Nota |
|---|---|---|
| **Buscar** un elemento | `document.querySelector('#titulo')` | Usa selectores de CSS: `#id`, `.clase`, `etiqueta`. Devuelve el **primero** (o `null` si no existe) |
| Buscar **todos** | `document.querySelectorAll('li')` | Devuelve una lista; se recorre con `.forEach(function (li) { … })` |
| **Leer / cambiar** el texto | `el.textContent = 'Hola'` | ✅ Seguro: lo escrito se muestra como **texto** |

| **Cambiar atributos** | `el.setAttribute('aria-live', 'polite')` · `el.hidden = true` | `hidden` oculta el elemento (pon `[hidden] { display: none !important; }` en tu CSS: ya lo hizo el Paso 0) |

### ⚠️ `textContent` vs `innerHTML` (la regla de seguridad de la semana)
`innerHTML` **interpreta** lo que recibe como HTML. Si el dato viene de una persona (un nombre en un formulario), puede traer una etiqueta maliciosa y **ejecutarla**: es un ataque **XSS** (*OWASP Top 10, A03 Inyección*). **Regla del curso: todo lo que venga de una persona se escribe con `textContent`.** Tus autoverificadores incluso lo comprueban.

### 2.4 · Eventos: la página escucha y responde
Un **evento** es algo que ocurre en la página: un clic, una tecla, un cambio en un campo *(MDN, Introducción a los eventos)*. Con `addEventListener` le dices al navegador **qué función ejecutar** cuando ocurra. Para cada interacción hazte **tres preguntas**: **① ¿en qué elemento?** → **② ¿qué evento?** → **③ ¿qué hago cuando ocurra?**

```javascript
boton.addEventListener('click', function (e) {
  contador.textContent = Number(contador.textContent) + 1;
});
```

| Evento | Cuándo ocurre | Úsalo para… |
|---|---|---|
| `click` | Clic o Enter/Espacio sobre un botón | Sumar, abrir, guardar |
| `input` | **Cada vez** que cambia el texto de un campo | Buscadores y totales en vivo |
| `change` | Al elegir otra opción de un `<select>` | Recalcular |
| `keydown` | Al presionar una tecla (`e.key` dice cuál: `'Escape'`, `'Enter'`) | Atajos |
| `submit` | Al enviar un formulario | Validar (sesión 2) |

- **Pasa la función, no la llames:** `addEventListener('input', filtrar)` ✅ · `addEventListener('input', filtrar())` ❌ (esto la ejecutaría **ya**, no al escribir).
- **`addEventListener` en lugar de `onclick="…"`:** separa el JavaScript del HTML y permite varios manejadores.
- **El objeto del evento (`e`)** trae datos: `e.key` (qué tecla), `e.target` (quién lo provocó).
- **Una función que “se acuerda”** (como el contador de tu predicción 2) es justo lo que es un manejador de eventos: se crea ahora y se ejecuta después, cuantas veces haga falta.

### 🔎 Cómo se hace un **buscador** (lo vas a construir hoy)
1. Escuchar el evento `input` del campo de búsqueda.
2. Leer lo escrito (`campo.value`) y **normalizarlo**: minúsculas y sin acentos, para que «Café» = «cafe» = «CAFE» → `texto.toLowerCase().normalize('NFD').replace(/[\u0300-\u036f]/g, '')`.
3. Recorrer las filas: si el texto de la fila **incluye** lo buscado, se muestra; si no, `fila.hidden = true`.
4. Contar las visibles y escribir **«X de N registros»**.
5. Accesibilidad: el campo necesita su **`<label>`** y el contador va en una región con **`aria-live="polite"`** para que un lector de pantalla lo anuncie *(WCAG 4.1.3 Mensajes de estado)*.

### 2.1 · Formularios: validar con respeto
El navegador ya sabe validar (`required`, `min`, `max`, `pattern`, `type="email"`…), pero sus globos de error son **genéricos** y cambian de navegador en navegador. Con JavaScript tú **controlas el mensaje** *(MDN, Validación de formularios y API de validación de restricciones)*.

**El ciclo de un formulario con validación propia:**
```
la persona escribe → pulsa «Enviar» (evento submit)
   → e.preventDefault()  (evita que la página se recargue)
   → por cada campo: ¿campo.validity.valid?
        no → escribir el mensaje junto al campo + aria-invalid + aria-describedby
        sí → quitar el mensaje
   → si hubo errores: llevar el foco al primer campo con error (focus) y terminar
   → si todo está bien: guardar / confirmar / limpiar (reset)
```

| Pieza | Qué hace |
|---|---|
| `<form novalidate>` (o `form.noValidate = true`) | Apaga los globos del navegador para mostrar **los tuyos** |
| `form.addEventListener('submit', function (e) { e.preventDefault(); … })` | Detiene el envío que recarga la página |
| `campo.validity.valid` | `true` si el campo cumple **todas** sus reglas |
| `validity.valueMissing` · `patternMismatch` · `rangeUnderflow` · `rangeOverflow` · `typeMismatch` | **Por qué** falló: vacío · formato · menor que `min` · mayor que `max` · tipo (correo) |
| `campo.setAttribute('aria-invalid', 'true')` | Avisa a los lectores de pantalla que el campo tiene un error |
| `campo.setAttribute('aria-describedby', 'error-nombre')` | Enlaza el campo con su mensaje, que se lee al enfocarlo |
| `campo.focus()` | Lleva al teclado al primer campo con error *(WCAG 2.4.3 Orden del foco)* |
| `formulario.reset()` | Limpia el formulario después de guardar |

**Un buen mensaje de error** *(WCAG 3.3.1 Identificación de errores y 3.3.2 Etiquetas o instrucciones; Nielsen #9: ayudar a reconocer y corregir errores)*: está **en español**, dice **qué hacer** («Escribe una edad entre 1 y 99») y se queda **junto al campo**. Y el mensaje de éxito va en una región `role="status"` para que también lo lea un lector de pantalla.

> 🛡️ **Seguridad:** validar en el navegador mejora la experiencia, **no protege** el sistema: alguien puede saltarse tu JavaScript. La validación que cuenta es la del **servidor** (semanas 8 y 9), *OWASP Input Validation Cheat Sheet*. Hoy practicas la parte de la experiencia de la persona usuaria.

### 2.2 · Números, formato y memoria del navegador
Lo que escribe una persona en un campo **siempre llega como texto**, aunque sea `3`. Por eso `'2' + '3'` da `'23'`, no `5`. Para calcular hay que **convertir** primero.

| Pieza | Qué hace | Ejemplo |
|---|---|---|
| `Number(texto)` | Convierte texto a número | `Number('3')` → `3` · `Number('abc')` → `NaN` |
| `Number.isInteger(n)` | ¿Es un entero? | `Number.isInteger(2.5)` → `false` |
| `n.toFixed(2)` | Redondea a 2 decimales (devuelve texto) | `(32.5 * 3).toFixed(2)` → `'97.50'` |
| `new Intl.NumberFormat('es-MX', { style: 'currency', currency: 'MXN' })` | Da formato de pesos con `.format(n)` | `.format(170)` → `$170.00` |
| `JSON.stringify(objeto)` · `JSON.parse(texto)` | Convierte un objeto o arreglo a **texto** y de regreso | `JSON.stringify([{a: 1}])` |
| `localStorage.setItem(clave, texto)` · `getItem(clave)` | Memoria del navegador (solo guarda **texto**) y **se conserva** al recargar | `localStorage.setItem('erp_form', JSON.stringify(lista))` |
| `tabla.createTBody().insertRow()` · `fila.insertCell()` | Arma filas y celdas | `fila.insertCell().textContent = dato` |
| `campo.name` · `campo.value` · `formulario.elements` | Recorre los campos de **cualquier** formulario | `Array.from(formulario.elements)` |

**El camino de un dato que se guarda y se recupera:**
```
campo.value (texto) ─► objeto { nombre: 'Café', precio: '85' } ─► se agrega a un ARREGLO
   ─► JSON.stringify(arreglo) ─► localStorage.setItem('erp_form', texto)      ← se guarda
...al recargar la página...
localStorage.getItem('erp_form') ─► JSON.parse(texto) ─► arreglo ─► se dibuja la tabla   ← se recupera
```

- **`localStorage` puede fallar** (modo privado, cuota llena): envuelve las lecturas y escrituras en `try/catch` y ten un valor de respaldo.
- **Para guardar registros:** un **arreglo de objetos** en una sola clave por formulario; cada campo se guarda por su `name` (por eso la semana 3 pedía `name` en todos los campos).
- 🛡️ **Nunca** pongas lo que escribió una persona en la tabla con `innerHTML`. Con `insertCell().textContent = …` una etiqueta `<img onerror=…>` se muestra como texto y **no se ejecuta**.
- ⚠️ `localStorage` **no es seguro** para datos sensibles (contraseñas, tarjetas): cualquiera con acceso al navegador puede leerlo. En la semana 8 los datos pasarán a una hoja de cálculo con credenciales protegidas.
